```python
def intersection(first, second):
    first_size = len(first) 
    second_size = len(second)
    values = []
    if first_size <= second_size:
        size = first_size
        data = second
        item = first
    else:
        size = second_size
        data = first
        item = second
        
    for i in range(size):
        if item[i] in data:
            values.append(item[i])    
                
    return values
        
def main(first, second):
    if not first and not second:
        print("Both the arrays are empty!")
    elif not second or not first:
        print("There are no intersection because one of the array is empty!")
    else:
        values = intersection(first, second)
        if values:
            size = len(values)
            if size == len(first) and size == len(second): 
                print("All the items in both array intersect!")
            else:
                print(values)
        else:
            print(f"There are no intersection between arrays {first} & {second}!")
         
```
