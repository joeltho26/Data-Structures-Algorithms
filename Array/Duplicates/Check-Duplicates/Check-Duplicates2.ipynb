```python
def check_duplicates(numbers, value):
    flag= False
    if value in numbers:
        flag = True if numbers.count(value) > 1 else False
        if flag:
            print(f"{value} duplicates found!")
    else:
        print(f"{value} not found in array!")
        return flag
    
    if not flag:
        print(f"{value} has no duplicates!")
    return flag

def main(numbers, values):
    if values or isinstance(values, int):
        if isinstance(values, list):
            for i in values:
                print(check_duplicates(numbers, i))
        else:
            print(check_duplicates(numbers, values))
```

```python
if __name__ == "__main__":
    numbers = [17, 4, 1, 20, 17, 9, 23, 18, 34, -15, -10]
    values = [18, 17, 0, 2]
    main(numbers, values)
```
