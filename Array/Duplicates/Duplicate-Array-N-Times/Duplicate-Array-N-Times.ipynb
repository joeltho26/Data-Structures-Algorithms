```python
def multiply_array(numbers, multiplier):
    return numbers * multiplier

def main(numbers, multiplier):
    if isinstance(multiplier, list):
        for i in multiplier:
            print(multiply_array(numbers, i))
    else:
        print(multiply_array(numbers, multiplier))
```

```python
numbers = [1, 60, 8, 2, 89, 3, 12]
multiplier = 2
main(numbers, multiplier)
```

```python
numbers = [1, 60, 8, 2, 89, 3, 12]
multiplier = [2, 3, 5]
main(numbers, multiplier)
```
