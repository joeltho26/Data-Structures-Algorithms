```python
def count_duplicates(numbers, value):
    if value in numbers:
        if numbers.count(value) > 1:
            print(f"{value} has duplicates!")
        else:
            print(f"{value} has no duplicates!")
        return numbers.count(value)
    else:
        print(f"{value} not found in array!")
        return 0

def main(numbers, values):
    if values or isinstance(values, int):
        if isinstance(values, list):
            for i in values:
                print("Count of {0}: {1}".format(i, count_duplicates(numbers, i)))
        else:
            print("Count of {0}: {1}".format(values, count_duplicates(numbers, values)))
```

```python
if __name__ == "__main__":
    numbers = [17, 4, 1, 20, 17, 9, 23, 18, 18, 17, 34, -15, -10]
    values = [18, 17, 0, 2]
    main(numbers, values)
```
