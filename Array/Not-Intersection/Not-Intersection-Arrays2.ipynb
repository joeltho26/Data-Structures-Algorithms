```python
def not_intersection(first, second):
    data = first.symmetric_difference(second)       
    return list(data)
        
def main(first, second):
    if not first and not second:
        print("The arrays are empty!")
    elif first and not second:
        print(first)
    elif second and not first:
        print(second)
    else:
        data = not_intersection(set(first), set(second))
        if data:
            print(data)
        else:
            print(f"All Items in both the arrays {first} & {second} intersect!")
```

```python
if __name__ == "__main__":
    first = [17, 4, 1, 20, 9]
    second = [1, 4, 8, 3, 87, 90, 17]
    main(first, second)
```
