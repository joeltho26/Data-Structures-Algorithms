```python
from copy import deepcopy

def not_intersection(first, second):
    first_size = len(first) 
    second_size = len(second)
    deleted = []
    if first_size <= second_size:
        size = first_size
        data = second
        item = first
    else:
        size = second_size
        data = first
        item = second
        
    for i in range(size):
        if item[i] in data:
            deleted.append(item[i])
            data.remove(item[i])
        
    for i in deleted:
        item.remove(i)     
                
    return data, item
        
def main(first, second):
    arr1, arr2 = deepcopy(first), deepcopy(second)
    if not first and not second:
        print("The arrays are empty!")
    elif first and not second:
        print(first)
    elif second and not first:
        print(second)
    else:
        first, second = not_intersection(first, second)
        data = first+second
        if data:
            print(data)
        else:
            print(f"All Items in both the arrays {arr1} & {arr2} intersect!")
```
