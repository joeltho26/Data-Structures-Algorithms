```python
def two_sum(nums: list[int], target: int) -> list[int]:
    for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            if (nums[i] + nums[j]) == target:
                return [i,j]
        
def main(nums, value):
    print(two_sum(nums, value))
```

```python
if __name__ == "__main__":
    nums = [2,7,11,15]
    target = 9
    main(nums, target)
```
