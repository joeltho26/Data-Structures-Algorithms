```python
from math import ceil, floor
def find_median_sorted_arrays(nums1: list[int], nums2: list[int]) -> float:
    updated_list = sorted(nums1+nums2)
    size = len(updated_list)
    mid_index = size//2
    if size%2 == 0:
        return (updated_list[int(mid_index-1)] + updated_list[int(mid_index)])/2
    else:
        return updated_list[int(mid_index)]
        
def main(nums1, nums2):
    print(find_median_sorted_arrays(nums1, nums2))
```

```python
if __name__ == "__main__":
    nums1 = [1,3] 
    nums2 = [2]
    main(nums1, nums2)
```
