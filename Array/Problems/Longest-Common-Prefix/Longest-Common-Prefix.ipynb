```python
def longestCommonPrefix(strs: list[str]) -> str:
        size = float('inf')
        data = None
        for i in strs:
            length = len(i)
            if length < size:
                size = length
                data = i

        res = ""
        for j in data:
            for index, val in enumerate(strs):
                if val.startswith(j):
                    val = val[1:]
                    if val:
                        strs[index] = val
                    continue
                else:
                    return res
                
            res += j

        return res
```

```python
strs = ["flower","flow","flight"]
print(longestCommonPrefix(strs))
```

```python
strs = ["dog","racecar","car"]
print(longestCommonPrefix(strs))
```
