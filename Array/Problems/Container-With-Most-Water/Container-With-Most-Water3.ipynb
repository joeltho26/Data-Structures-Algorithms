```python
def max_area(height: list[int]) -> int:
    left = 0 
    right = len(height) - 1
    max_w = 0

    while left < right:
        h_left, h_right = height[left], height[right]
        min_h = min(h_left, h_right)
        max_w = max(max_w, min_h * (right - left))

        if h_left < h_right:
            while left < right and height[left] <= min_h:
                left += 1
        else:
            while left < right and height[right] <= min_h:
                right -= 1
    return max_w

def main(height):
    print(max_area(height))
```

```python
if __name__ == "__main__":
    height = [1,8,6,2,5,4,8,3,7]
    main(height)
    
```
