```python
def divide_array_by_index(values, index):
    size = len(values)
    if index < size - 1:
        first = values[:index+1]
        second = values[index+1:]
        return first, second 
    else:
        if index == size - 1:
            print(f"Index {index} does not split the array!")
        else:
            print(f"Invalid index {index}")
            
        return None, None

def main(numbers, indexes):
    if indexes or isinstance(indexes, int):
        if isinstance(indexes, list):
            for i in indexes:
                first, second = divide_array_by_index(numbers, i)
                if first is not None and second is not None:
                    print(f"Divided array at {i}")
                    print(first, second)
        else:
            first, second = divide_array_by_index(numbers, indexes)
            print(f"Divided array at {indexes}")
```

```python
if __name__ == "__main__":
    numbers = [17, 4, 1, 20, 17, 9, 23, 18, 34, -15, -10]
    indexes = [6, 0, 11, 10]
    main(numbers, indexes)
```
