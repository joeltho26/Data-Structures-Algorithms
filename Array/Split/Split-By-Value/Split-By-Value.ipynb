```python
def divide_array_by_value(values, value):
    if value in values:
        for i in range(len(values)):
            if values[i] == value:
                first = values[:i+1]
                second = values[i+1:]
                if not second:
                    print(f"The array cannot be split with value {value}")
                    return None, None
                break
        return first, second
    else:
        print(f"value {value} not found in the array!") 
        return None, None

def main(numbers, values):
    if values or isinstance(values, int):
        if isinstance(values, list):
            for i in values:
                first, second = divide_array_by_value(numbers, i)
                if first is not None and second is not None:
                    print(f"Divided array at {i}")
                    print(first, second)
        else:
            first, second = divide_array_by_value(numbers, values)
            if first is not None and second is not None:
                print(f"Divided array at {values}")
                print(first, second)
```

```python
if __name__ == "__main__":
    numbers = [17, 4, 1, 20, 17, 9, 23, 18, 34, -15, -10]
    indexes = [17, -10, 11, 23]
    main(numbers, indexes)
```
