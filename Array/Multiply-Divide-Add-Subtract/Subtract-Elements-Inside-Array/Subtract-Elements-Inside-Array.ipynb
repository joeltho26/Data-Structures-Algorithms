```python
def subtract_array_based_on_value(numbers, subtract):
    return list(map(lambda x: x-subtract, numbers))

def main(numbers, subtract):
    if isinstance(subtract, list):
        for i in subtract:
            print(subtract_array_based_on_value(numbers, i))
    else:
        print(subtract_array_based_on_value(numbers, subtract))
        
```

```python
numbers = [1, 60, 8, 1, 89, 3, 12]
subtract = 2
main(numbers, subtract)
```

```python
numbers = [1, 60, 8, 2, 89, 3, 12]
subtract = [2, 3, 5]
main(numbers, subtract)
```
