```python
def multiply_array_based_on_value(numbers, multiplier):
    return list(map(lambda x: x* multiplier, numbers))

def main(numbers, multiplier):
    if isinstance(multiplier, list):
        for i in multiplier:
            print(multiply_array_based_on_value(numbers, i))
    else:
        print(multiply_array_based_on_value(numbers, multiplier))
        
```

```python
numbers = [1, 60, 8, 1, 89, 3, 12]
multiplier = 2
main(numbers, multiplier)
```

```python
numbers = [1, 60, 8, 2, 89, 3, 12]
multiplier = [2, 3, 5]
main(numbers, multiplier)
```
