```python
def add_array_based_on_value(numbers, add):
    return list(map(lambda x: x+add, numbers))

def main(numbers, add):
    if isinstance(add, list):
        for i in add:
            print(add_array_based_on_value(numbers, i))
    else:
        print(add_array_based_on_value(numbers, add))
        
```

```python
numbers = [1, 60, 8, 1, 89, 3, 12]
add = 2
main(numbers, add)
```

```python
numbers = [1, 60, 8, 2, 89, 3, 12]
add = [2, 3, 5]
main(numbers, add)
```
