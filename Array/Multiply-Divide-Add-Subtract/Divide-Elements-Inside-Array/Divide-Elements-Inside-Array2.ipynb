```python
def division(val, divisor):
    return round(val/divisor, 2)

def divide_array_based_on_value(numbers, divisor):
    return list(map(lambda x: division(x, divisor), numbers))

def main(numbers, divisor):
    if isinstance(divisor, list):
        for i in divisor:
            print(divide_array_based_on_value(numbers, i))
    else:
        print(divide_array_based_on_value(numbers, divisor))
        
```

```python
numbers = [1, 60, 8, 1, 89, 3, 12]
divisor = 2
main(numbers, divisor)
```

```python
numbers = [1, 60, 8, 2, 89, 3, 12]
divisor = [2, 3, 5]
main(numbers, divisor)
```
