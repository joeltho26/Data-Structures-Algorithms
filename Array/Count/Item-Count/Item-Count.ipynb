```python
def item_count(numbers, value):
    if value in numbers:
        return numbers.count(value)
    else:
        print(f"{value} not found in array!")
        return
    
def main(numbers, values):
    if values or isinstance(values, int):
        if isinstance(values, list):
            for i in values:
                value = item_count(numbers, i)
                if value:
                    print(f"The count of {i} is: {value}")
        else:
            value = item_count(numbers, values)
            if value:
                print(f"The count of {values} is: {value}")
    else:
        print("No values to check item count")
```

```python
if __name__ == "__main__":
    numbers = [17, 4, 1, 20, 17, 9, 23, 18, 34, -15, -10]
    values = [18, 17, 0, 2]
    main(numbers, values)
```
