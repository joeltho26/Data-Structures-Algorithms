```python
def delete_by_value(values, value):
    if value in values:
        for i in range(len(values)):
            if values[i] == value:
                value = values.pop(i)
                break 
    else:
        print(f"Value {value} not found!")
        value = None
    return value, values

def main(numbers, values):
    if values or isinstance(value, int):
        if isinstance(values, list):
            for i in values:
                value, numbers = delete_by_value(numbers, i)
                if value:
                    print(f"The deleted value is: {value}")
                    print(f"Remaining values after deletion: {numbers}")
        else:
            value, numbers = delete_by_value(numbers, values)
            if value:
                print(f"The deleted value is: {value}")
                print(f"Remaining values after deletion: {numbers}")

    else:
        print("No values available for deletion!")
        
    return numbers
```
