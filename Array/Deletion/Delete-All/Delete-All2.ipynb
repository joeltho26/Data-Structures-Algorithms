```python
def delete(numbers):
    value = numbers
    remaining = []
    return value, remaining

def main(numbers):
    value, numbers = delete(numbers)
    print(f"The deleted values are: {value}")
    print(f"Remaining values after deletion: {numbers}")
    return numbers
```

```python
if __name__ == "__main__":
    numbers = [17, 4, 1, 20, 17, 9, 23, 18, 34, -15, -10]
    print(f"Values before deletion: {numbers}")
    numbers = main(numbers)
    print(numbers)
```
