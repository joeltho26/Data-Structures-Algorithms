```python
def delete_at_end(numbers):
    return numbers[-1], numbers[:-1]

def main(numbers):
    if numbers:
        value, numbers = delete_at_end(numbers)
        print(f"The deleted value is: {value}")
        print(f"Remaining values after deletion: {numbers}")
        return numbers
    else:
        print("The array is empty!")
        
```

```python
if __name__ == "__main__":
    numbers = [17, 4, 1, 20, 17, 9, 23, 18, 34, -15, -10]
    print(f"Values before deletion: {numbers}")
    numbers = main(numbers)
    print(numbers)
```
