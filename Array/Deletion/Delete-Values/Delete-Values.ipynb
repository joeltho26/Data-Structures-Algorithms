```python
def delete_values(values, values_to_delete):
    values_deleted = []
    if values_to_delete:
        if isinstance(values_to_delete, list):
            for i in values_to_delete:
                if i in values:                            
                    values.remove(i)
                    values_deleted.append(i)
                else:
                    print(f"Value {i} not found!")
        else:
            if values_to_delete in values:
                values_deleted.append(values_to_delete)
                values.remove(values_to_delete)
            else:
                print(f"Value {values_to_delete} not found!")
                
    return values_deleted, values

def main(numbers, values):
    if values or isinstance(values, int):
        values, numbers = delete_values(numbers, values)
        if values:
            if len(values) < 2:
                print(f"The deleted value is: {values[0]}")
            else:
                print(f"The deleted values are: {values}")
            print(f"Remaining values after deletion: {numbers}")
        else:
            print("No values were deleted!")
    else:
        print("No values available for deletion!")
        
    return numbers
    
if __name__ == "__main__":
    numbers = [17, 4, 1, 20, 17, 9, 23, 18, 34, -15, -10]
    value = [17, 48]
    print(f"Values before deletion: {numbers}")
    numbers = main(numbers, value)
    print(numbers)    
```
