```python
def subtract_arrays(values, to_be_deleted):
    deleted = None
    not_found = None
    if isinstance(to_be_deleted, list):
        deleted = []
        not_found = []
        for i in to_be_deleted:
            if i in values:
                values.remove(i)
                deleted.append(i)
            else:
                not_found.append(i)
    else:
        if to_be_deleted in values:
            values.remove(i)
            deleted = to_be_deleted
        else:
            not_found.append(i)
            
    return deleted, not_found, values

def main(values, to_be_deleted):
    size = len(values)
    deleted, not_found, values = subtract_arrays(values, to_be_deleted)
    if not values:
        print("All the items in the array are deleted:")
    elif size == len(values):
        print("No items in the array are deleted:")
    
    print(f"Items deleted from the array: {deleted}")
    print(f"Items not found in the array to delete: {not_found}")
    print(f"Items in the array: {values}")
    
```

```python
if __name__ == "__main__":
    numbers = [17, 4, 1, 20, 17, 9, 23, 18, 34, -15, -10]
    to_be_deleted = [1, 4, 8, 1, 23, 3, 87, 90, 17, 54, 12, 60, 89]
    main(numbers, to_be_deleted)
```
