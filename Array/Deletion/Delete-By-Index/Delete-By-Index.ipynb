```python
def delete_at(values, index):
    size = len(values)
    if index < size:
        value = values.pop(index)
    else:
        value = None
        print(f"Invalid Index {index}!")
    return value, values

def main(numbers, index):
    if index or index > -1:
        if isinstance(index, list):
            for i in index:
                value, numbers = delete_at(numbers, i)
                if value is not None:
                    print(f"The deleted value via index {index} is: {value}")
                    print(f"Remaining values after deletion: {numbers}")
        else:
            value, numbers = delete_at(numbers, index)
            if value is not None:
                print(f"The deleted value via index {index} is: {value}")
                print(f"Remaining values after deletion: {numbers}")
    else:
        print("No index available for deletion!")
        
    return numbers
```

```python
if __name__ == "__main__":
    numbers = [17, 4, 1, 20, 17, 9, 23, 18, 34, -15, -10]
    index = [8, 20]
    print(f"Values before deletion: {numbers}")
    numbers = main(numbers, index)
    print(numbers)
```
