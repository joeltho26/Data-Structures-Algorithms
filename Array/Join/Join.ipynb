```python
def add(values):
    data = []
    for i in values:
        data.extend(i)
    return data

def main(values):
    print(add(values))
```

```python
if __name__ == "__main__":
    numbers = [[1,4,8,1], [23,3,87,90], [7,54,12,60,89]]
    main(numbers)
```
