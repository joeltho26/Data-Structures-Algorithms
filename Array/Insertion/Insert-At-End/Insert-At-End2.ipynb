```python
def insert_at_end(numbers, to_insert):
    numbers += [to_insert]
    return numbers

def main(numbers, to_insert):
    if to_insert:
        if isinstance(to_insert, list):
            for i in to_insert:
                numbers = insert_at_end(numbers, i)
                print(f"Inserted element {i} at the beginning!")
        else:
           numbers = insert_at_end(numbers, to_insert)
           print(f"Inserted element {to_insert} at the beginning!")
    else:
        print("No values available to insert!")
    return numbers
        
```

```python
if __name__ == "__main__":
    numbers = [17, 4, 1, 20, 17, 9, 23, 18, 34, -15, -10]
    to_insert = [4, 7, 9, 43]
    print(f"Values before insertion: {numbers}")
    numbers = main(numbers, to_insert)
    print(f"Values after insertion: {numbers}")
```
