```python
def insert_by_value(numbers, value, insert_value):
    index = numbers.index(value)
    numbers.insert(index, insert_value)
    print(f"The value {insert_value} is inserted at index {index}")
    return numbers

def main(numbers, values, insert_value):
    if values:
        if isinstance(values, list):
            for i in values:
                if i in numbers:
                    numbers = insert_by_value(numbers, i, insert_value)
                else:
                    print(f"No match found for {i}")
        else:
            if values in numbers:  
                numbers = insert_by_value(numbers, values, insert_value)
            else:
                print(f"No match found for {values}")
    else:
        print("No values available to insert!")
        
    return numbers
```

```python
if __name__ == "__main__":
    numbers = [17, 4, 1, 20, 17, 9, 23, 18, 34, -15, -10]
    values = [4, 0, -10, 12]
    insert_value = 100
    print(f"Values before insertion: {numbers}")
    numbers = main(numbers, values, insert_value)
    print(f"Values after insertion: {numbers}")
```
