```python
def insert_by_index(numbers, index, value):
    size = len(numbers)
    if index < size:
        numbers = numbers[:index] + [value] + numbers[index:]
        print(f"Inserted element {value} at index {index}!")
    elif index == size:
        numbers += [value]
    else:
        print("Invalid Index!")
    return numbers

def main(numbers, indexes, value):
    if indexes:
        if isinstance(indexes, list):
            for i in indexes:
                numbers = insert_by_index(numbers, i, value)
        else:
            numbers = insert_by_index(numbers, indexes, value)
    else:
        print("No values available to insert!")
        
    return numbers
```

```python
if __name__ == "__main__":
    numbers = [17, 4, 1, 20, 17, 9, 23, 18, 34, -15, -10]
    indexes = [4, 0, 43]
    value = 100
    print(f"Values before insertion: {numbers}")
    numbers = main(numbers, indexes, value)
    print(f"Values after insertion: {numbers}")
```
