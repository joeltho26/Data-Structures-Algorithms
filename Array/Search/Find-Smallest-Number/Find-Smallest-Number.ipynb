```python
def smallest_number(numbers):
    smallest = float('inf')
    for i in numbers:
        if i < smallest:
            smallest = i
    return smallest

def main(numbers):
    print(smallest_number(numbers))
```

```python
if __name__ == "__main__":
    numbers = [17, 4, 1, 20, 17, 9, 23, 18, 34, -15, -10]
    main(numbers)
```
