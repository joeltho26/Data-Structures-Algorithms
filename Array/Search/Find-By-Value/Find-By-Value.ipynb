```python
def search(values, value):
    flag = False
    if value in values:
        flag=True
    return flag

def main(numbers, values):
    if values or isinstance(values, int):
        if isinstance(values, list):
            for i in values:
                print(search(numbers, i)) 
        else:
            print(search(numbers, values))
```

```python
if __name__ == "__main__":
    numbers = [17, 4, 1, 20, 17, 9, 23, 18, 34, -15, -10]
    values = [17, 34, 0, 15]
    main(numbers, values)
```
