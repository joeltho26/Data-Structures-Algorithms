```python
def highest_number(numbers):
    highest = float('-inf')
    for i in numbers:
        if i > highest:
            highest = i
    return highest

def main(numbers):
    print(highest_number(numbers))
```

```python
if __name__ == "__main__":
    numbers = [17, 4, 1, 20, 17, 9, 23, 18, 34, -15, -10]
    main(numbers)
```
