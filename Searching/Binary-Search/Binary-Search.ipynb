```python
def binary_search(numbers, to_find):
    left = 0
    size = right = len(numbers)
    
    while left <= right:
        mid_index = (left+right)//2
        mid_val = numbers[mid_index]
        if to_find == mid_val:
            return mid_index
        
        if to_find > mid_val:
            left = mid_index + 1
        
        if to_find < mid_val:
            right = mid_index - 1
            
    return -1
        
def main(numbers, to_find):
    index = binary_search(numbers, to_find)
    if index >= 0:
        print(f"The index for {to_find} is {index}")
    else:
        print(f"No match found for {to_find}!")
    
```
