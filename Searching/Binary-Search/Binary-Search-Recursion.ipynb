```python
def binary_search_recursion(numbers, to_find, left, right):
    
    if left > right or left < 0 or right > len(numbers):
        return -1

    while left <= right:
        mid_index = (left+right)//2
        mid_val = numbers[mid_index]
        if to_find == mid_val:
            return mid_index
        
        if to_find > mid_val:
            left = mid_index + 1
        
        if to_find < mid_val:
            right = mid_index - 1
            
    return binary_search_recursion(numbers, to_find, left, right)
        
def main(numbers, to_find, left, right):
    index = binary_search_recursion(numbers, to_find, left, right)
    if index >= 0:
        print(f"The index for {to_find} is {index}")
    else:
        print(f"No match found for {to_find}!")
    
```
