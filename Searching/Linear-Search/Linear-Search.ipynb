```python
def linear_search(numbers, to_find):
    for i in range(len(numbers)):
        if numbers[i] == to_find:
            return i
        
def main(numbers, to_find):
    index = linear_search(numbers, to_find)
    if index:
        print(f"The index for {to_find} is {index}")
    else:
        print("No match found!")
```

```python
if __name__ == "__main__":
    numbers = [17, 4, 1, 20, 17, 9, 23, 18, 34, -15, -10]
    to_find = 18
    main(numbers, to_find)
```
