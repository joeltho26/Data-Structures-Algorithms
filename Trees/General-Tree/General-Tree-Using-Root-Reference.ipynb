```python
import json
from pathlib import Path
import os
import re
from json import JSONDecodeError
from typing import Optional

class GeneralTree:
    def __init__(self, data: str|None=None):
        self.data = data
        self.children = []
        self.parent = None

    def add_child(self, root, data):      
        if not root:
            return data
        
        root.children.append(data)
        data.parent = root

    def get_level(self):
        level = 0
        itr = self.parent
        while itr:
            itr = itr.parent
            level += 1
        return level

    def print_tree(self, root):
        if not root:
            print("General Tree is empty!")
            return
        
        spaces = " " * root.get_level() * 2
        prefix = spaces + "|--" if root.parent else ""
        print(prefix+root.data)
        if root.children:
            for child in root.children:
                self.print_tree(child)
                
    def depth_of(self, root, data, flag=False):
        depth = None
        if not root:
            print("General Tree is empty!")
            return depth
        
        if root.data == data:
            return root.get_level()
        
        if root.children:
            for child in root.children:
                depth = self.depth_of(child, data, flag=True)
                if depth is not None:
                    return depth
                
        if not flag:
            print(f"No match found for {data}")
                    
        return depth
    
    def get_depth(self, root):
        if not root:
            print("General Tree is empty!")
            return 0
        
        depth = 0
        if root.children:
            for child in root.children:
                depth = max(self.get_depth(child), depth)
            
        return depth + 1
    
    def get_height(self, root):
        if not root:
            print("General Tree is empty!")
            return 0
        
        height = 0
        if root.children:
            for child in root.children:
                height = max(self.get_height(child), height)
            
        return height + 1

    def height_of(self, root, data, flag=False):
        height = None
        if not root:
            print("General Tree is empty!")
            return height
        
        if root.data == data:
            return self.get_height(root) - 1
        
        if root.children:
            for child in root.children:
                height = self.height_of(child, data, flag=True)
                if height is not None:
                    return height
                
        if not flag:
            print(f"No match found for {data}")    
        
        return height
                    
    def get_root(self, root, rotation: Optional[bool]=False):
        if root:
            if root.parent is None:
                return root.data
            else:
                if not rotation:
                    print("Not a root node!")
                return
        else:
            if not rotation:
                print("General Tree is empty!")
            return
    
    def get_height(self, data):
        return self.get_depth(data)

    def get_diameter(self, root):
        if not root:
            print("General Tree is empty!")
            return 0
        
        left, right = 0, 0
        if root.children:
            for child in root.children:
                node = self.get_depth(child)
                if node > left:
                    right, left = left, node
                elif node > right:
                    right = node

        diameter = 0
        for child in root.children:
            diameter = max(diameter, self.get_diameter(child))
        return max(diameter, left+right+1)

    def get_diameter_of(self, root, data, flag=False):
        diameter = None
        if not root:
            print("General Tree is empty!")
            return diameter
        
        if root.data == data:
            return self.get_diameter(root)

        if root.children:
            for child in root.children:
                diameter = self.get_diameter_of(child, data, flag=True)
                if diameter is not None:
                    return diameter
                
        if not flag:
            print(f"No match found for {data}")
                
        return diameter 
    
    def is_root(self, root):
        if root:
            if root.parent is None:
                return True
    
        return False
                
    def count_items(self, root, count=0):
        if not root:
            print("General Tree is empty!")
            return 0
        
        if root.children:
            for child in root.children:
                count = self.count_items(child, count)
        return count + 1
            
    def get_leaf_nodes(self, root, leaves=None):
        if leaves is None:
            leaves = []
            
        if not root:
            print("General Tree is empty!")
            return
        
        if not root.children:
            leaves.append(root.data)
        
        if root.children:
            for child in root.children:
                self.get_leaf_nodes(child, leaves)
        return leaves
    
    def get_leaf_nodes_count(self, root):
        return len(self.get_leaf_nodes(root))
              
    def get_parent(self, root, data, is_found=None, flag=False):
        if not root:
            print("General Tree is empty!")
            return None
        
        if is_found is None:
            is_found = [False]
        
        parent = None
        if root.data == data:
            is_found[0] = True
            if not self.is_root(root):
                if root.parent:
                    return root.parent.data
            
            print(f"{data} is the root node and has no parent!")
            return
            
        if root.children:
            for child in root.children:
                parent = self.get_parent(child, data, is_found, flag=True)
                if parent or is_found[0]:
                    if parent is None:
                        return
                    return parent
                
        if not flag and not is_found[0]:
            print(f"No match found for {data}") 
            
        return parent
    
    def get_parents(self, root, data, parents=None, is_found=None, flag=False):
        if not root:
            print("General Tree is empty!")
            return []
        
        if parents is None:
            parents = []
            
        # mutable list/dictionary can have shared values across every level of recursion 
        if is_found is None:
            is_found = [False]
        
        if root.data == data:
            is_found[0] = True
            if not self.is_root(root):
                if root.parent:
                    parents.append(root.parent.data)
            else:  
                print(f"{data} is the root node and has no parent!") 
            
        if root.children:
            for child in root.children:
                self.get_parents(child, data, parents, is_found, flag=True)
                
        if not flag and not is_found[0]:
            print(f"No match found for {data}")
            
        if not parents:
            return
            
        return parents
    
    def get_children(self, root, data, is_found=None, flag=False):
        children = None
        if not root:
            print("General Tree is empty!")
            return children
        
        if is_found is None:
            is_found = [False]
        
        if root.data == data:
            is_found[0] = True
            if root.children:
                children = []
                for child in root.children:
                    children.append(child.data)
                return children
            else:
                print(f"Match found, but no child available for {data}!")               
                return 
                    
        if root.children:
            for child in root.children:
                children = self.get_children(child, data, is_found, flag=True)
                if children or is_found[0]:
                    if children is None:
                        return
                    return children
                
        if not flag and not is_found[0]:
            print(f"No match found for {data}") 
        
        return children
    
    def get_all_children(self, root, data, children=None, is_found=None, flag=False):
        if children is None:
            children = []
            
        # global keyword keeps the variable shared across every level of recursion but it stays the same throughtout even if we call the N number of times for a different use cases
        if is_found is None:
            is_found = [False]
            
        if not root:
            print("General Tree is empty!")
            return children
        
        if root.data == data:
            is_found[0] = True
            if root.children:
                for child in root.children:
                    children.append(child.data)
            else:
                print(f"Match found, but no child available for {data}!")
                    
        if root.children:
            for child in root.children:
                self.get_all_children(child, data, children, is_found, flag=True)
                
        if not flag and not is_found[0]:
            print(f"No match found for {data}") 
            
        if not children:
            return
        
        return children
    
    def get_grand_children(self, root, data, is_found=None, flag=False):
        grand_children = None
        if not root:
            print("General Tree is empty!")
            return 
        
        if is_found is None:
            is_found = [False]
        
        if root.data == data:
            is_found[0] = True
            if root.children:
                for child in root.children:
                    if child.children:
                        grand_children = []
                        for grand_child in child.children:
                            grand_children.append(grand_child.data)
                        return grand_children
                    else:
                        print(f"Match found, but no grand children available for {data}!")
                        return
            else:
                print(f"Match found, but no children and grand children available for {data}!")
                return
        
        if root.children:
            for child in root.children:
                grand_children = self.get_grand_children(child, data, is_found, flag=True)
                if grand_children or is_found[0]:
                    if grand_children is None:
                        return
                    return grand_children
                
        if not flag and not is_found[0]:
            print(f"No match found for {data}") 
        
        return grand_children
    
    def get_all_grand_children(self, root, data, grand_children=None, is_found=None, flag=False): 
        if grand_children is None:
            grand_children = []
            
        if is_found is None:
            is_found = [False]
            
        if not root:
            print("General Tree is empty!")
            return grand_children
    
        if root.data == data:
            is_found[0] = True
            if root.children:
                for child in root.children:
                    if child.children:
                        for grand_child in child.children:
                            grand_children.append(grand_child.data)
                            
                if not grand_children:
                    print(f"Match found, but no grand children available for {data}!")

            else:
                print(f"Match found, but no children and grand children available for {data}!")    
        
        
        if root.children:
            for child in root.children:
                self.get_all_grand_children(child, data, grand_children, is_found, flag=True)
                
        if not flag and not is_found[0]:
            print(f"No match found for {data}") 
            
        if not grand_children:
            return
        
        return grand_children
    
    def get_grand_parent(self, root, data, is_found=None, flag=False):
        grand_parent = None
        if not root:
            print("General Tree is empty!")
            return grand_parent
        
        if is_found is None:
            is_found = [False]
        
        if root.data == data:
            is_found[0] = True
            if root.parent:
                if root.parent.parent:
                    return root.parent.parent.data
                
                print(f"{data} has no grand parent!")
                return 
            
            print(f"{data} is a root node and has no parent and grand parent!")
            return
            
        if root.children:
            for child in root.children:
                grand_parent = self.get_grand_parent(child, data, is_found, flag=True)
                if is_found[0] or grand_parent:
                    if not grand_parent:
                        return
                    return grand_parent
                
        if not flag and not is_found[0]:
            print(f"No match found for {data}") 
            
        return grand_parent
    
    def get_all_grand_parent(self, root, data, grand_parents=None, is_found=None, flag=False):
        if grand_parents is None:
            grand_parents = []
            
        if is_found is None: 
            is_found = [False]
            
        if not root:
            print("General Tree is empty!")
            return grand_parents
    
        if root.data == data:
            is_found[0] = True
            if root.parent:
                if root.parent.parent:
                    grand_parents.append(root.parent.parent.data)
                else:
                    print(f"{data} has no grand parent!")
            else:
                print(f"{data} is a root node and has no parent and grand parent!")
            
        if root.children:
            for child in root.children:
                self.get_all_grand_parent(child, data, grand_parents, is_found, flag=True)
                
        if not flag and not is_found[0]:
            print(f"No match found for {data}") 
            
        if not grand_parents:
            return
            
        return grand_parents
        
    def get_children_count(self, root, data):
        children = self.get_children(root, data)
        if children:
            return len(children)
        else:
            return 0
    
    def get_grand_children_count(self, root, data):
        children = self.get_grand_children(root, data)
        if children:
            return len(children)
        else:
            return 0
    
    def get_count_key(self, root, data):
        count = 0
        if not root:
            print("General Tree is empty!")
            return 0
        
        if root.data == data:
            return 1
        
        if root.children:
            for child in root.children:
                count += self.get_count_key(child, data)
        return count
        
    def search(self, root, data, validate: Optional[bool]=False, flag=False):
        if not root:
            print("General Tree is empty!")
            return False
        
        if root.data == data:
            if not validate:
                print(f"Match found for {data}!")
            return True
        
        if root.children:
            for child in root.children:
                res = self.search(child, data, flag=True)
                if res:
                    return res
                
        if not flag:
            if not validate:
                print(f"No match found for {data}")
            return False
    
    def get_first_child(self, root, first_children=None):
        if first_children is None:
            first_children = []
        
        if not root:
            print("General Tree is empty!")
            return 
        
        if root.children:
            first_children.append(root.children[0].data)
                
        if root.children:
            for child in root.children:
                self.get_first_child(child, first_children)
                
        return first_children
    
    def get_first_child_of(self, root, data, first_children=None, flag=False):
        if first_children is None:                                     
            first_children = []
            
        if not root:
            print("General Tree is empty!")
            return 
        
        if root.data == data:
            if root.children:
                first_children.append(root.children[0].data)
                
        if root.children:
            for child in root.children:
                self.get_first_child_of(child, data, first_children, flag=True)
                
        if not flag and not first_children:
            print(f"{data} not found in the General Tree!")
            return
            
        return first_children

    def get_middle_child(self, root, middle_children=None):
        if middle_children is None:
            middle_children = []
        
        if not root:
            print("General Tree is empty!")
            return 
        
        if root.children:
            mid_index = len(root.children)//2
            middle_children.append(root.children[mid_index].data)
                
        if root.children:
            for child in root.children:
                self.get_middle_child(child, middle_children)
                
        return middle_children
    
    def get_middle_child_of(self, root, data, middle_children=None, flag=False): 
        if middle_children is None:
            middle_children = []
                   
        if not root:
            print("General Tree is empty!")
            return
        
        if root.data == data:
            if root.children:
                mid_index = len(root.children)//2
                middle_children.append(root.children[mid_index].data)
                
        if root.children:
            for child in root.children:
                self.get_middle_child_of(child, data, middle_children, flag=True)
                
        if not flag and not middle_children:
            print(f"{data} not found in the General Tree!")
            return
        
        return middle_children

    def get_last_child(self, root, last_children=None):
        if last_children is None:
            last_children = []
        
        if not root:
            print("General Tree is empty!")
            return
        
        if root.children:
            last_children.append(root.children[-1].data)
                
        if root.children:
            for child in root.children:
                self.get_last_child(child, last_children)
                
        return last_children
        
    def get_last_child_of(self, root, data, last_children=None, flag=False):
        if last_children is None:
            last_children = []
            
        if not root:
            print("General Tree is empty!")
            return
        
        if root.data == data:
            if root.children:
                last_children.append(root.children[-1].data)
                
        if root.children:
            for child in root.children:
                self.get_last_child_of(child, data, last_children, flag=True)
            
        if not flag and not last_children:
            print(f"{data} not found in the General Tree!")
            return
        
        return last_children
    
    def dfs_traversal(self, root, data=None):
        if data is None:
            data = []

        if not root:
            print("General Tree is empty!")
            return
        
        data.append(root.data)
        
        if root.children:
            for child in root.children:
                self.dfs_traversal(child, data)
                
        return data
    
    def bfs_traversal(self, root, data=None):
        if data is None:
            data = []
            
        if not root:
            print("General Tree is empty!")
            return
        
        q = [root]
        while q:
            for _ in range(len(q)):
                node = q.pop(0)
                data.append(node.data)
                
                if node.children:
                    for child in node.children:
                        q.append(child)
        return data
    
    def insert_sibling_by_value(self, root, sibling, value, is_inserted=None, is_found=None, flag=False):
        if not root:
            print("General Tree is empty!")
            return False
        
        if is_inserted is None and is_found is None:
            is_inserted = [False]
            is_found = [False] 
                
        if self.get_root(root, rotation=True) != sibling:  
            if root.children:
                values = [child.data for child in root.children]
                if value not in values:
                    if sibling in values:
                        is_inserted[0] = is_found[0] = True
                        node = GeneralTree(value)
                        root.children.append(node)
                        node.parent = root
                        if root.data == self.get_root(root, rotation=True):
                            return is_inserted[0]
                        return (is_inserted, is_found)
                else:
                    print(f"Value {value} already exists in the tree!")
                    is_found[0] = True
                    return (is_inserted, is_found)
        else:
            print(f"Match found for {self.get_root(root, rotation=True)}, but no parent available to add it as sibling")
            return is_inserted[0]
                
        if root.children:
            for child in root.children:
                result = self.insert_sibling_by_value(child, sibling, value, is_inserted, is_found, flag=True)
                if isinstance(result, tuple):
                    is_inserted, is_found = result
                    if is_inserted[0] or is_found[0]:
                        return is_inserted[0]
                
        if not flag and not is_found[0]:
            print(f"No match found for sibling {sibling}") 
            
        return is_inserted[0]
    
    def insert_sibling_by_value_for_all(self, root, sibling, value, is_inserted=None, is_found=None, flag=False):
        if not root:
            print("General Tree is empty!")
            return False
        
        if is_inserted is None:
            is_inserted = [False]
            is_found = [False]
        
        if self.get_root(root, rotation=True) != sibling:
            if root.children:
                values = [child.data for child in root.children]
                if value not in values:
                    if sibling in values:
                        is_inserted[0] = is_found[0] = True
                        node = GeneralTree(value)
                        root.children.append(node)
                        node.parent = root
                else:
                    print(f"Value {value} already exists in the tree!")
                    is_found[0] = True
        else:
            print(f"Match found for {self.get_root(root, rotation=True)}, but no parent available to add it as sibling")
            is_found[0] = True
                
        if root.children:
            for child in root.children:
                self.insert_sibling_by_value_for_all(child, sibling, value, is_inserted, is_found, flag=True)
                
        if not flag and not is_found[0]:
            print(f"No match found for sibling {sibling}") 
            
        return is_inserted[0]
    
    def insert_sibling_by_parent_value(self, root, parent, value, is_inserted=None, is_found=None, flag=False):
        if not root:
            print("General Tree is empty!")
            return False
        
        if is_inserted is None and is_found is None:
            is_inserted = [False]
            is_found = [False]

        if root.data == parent:
            is_found[0] = True
            values = [child.data for child in root.children]
            if value not in values:
                is_inserted[0] = True
                node = GeneralTree(value)
                root.children.append(node)
                node.parent = root
            else:
                print(f"Value {value} already exists in the tree!")

            if self.get_root(root, rotation=True) == parent:
                return is_inserted[0]
            return (is_inserted, is_found)
            
        if root.children:
            for child in root.children:
                result = self.insert_sibling_by_parent_value(child, parent, value, is_inserted, is_found, flag=True)
                if isinstance(result, tuple):
                    is_inserted, is_found = result
                if is_inserted[0] or is_found[0]:
                    return is_inserted[0]
            
        if not flag and not is_found[0]:
            print(f"No match found for parent {parent}") 
        
        return is_inserted[0]
    
    def insert_sibling_by_parent_value_for_all(self, root, parent, value, is_inserted=None, is_found=None, flag=False):
        if not root:
            print("General Tree is empty!")
            return False
        
        if is_inserted is None:
            is_inserted = [False]
            is_found = [False]
        
        if root.data == parent:
            values = [child.data for child in root.children]
            is_found[0] = True
            if value not in values:
                is_inserted[0] = True   
                node = GeneralTree(value)
                root.children.append(node)
                node.parent = root
            else:
                print(f"Value {value} already exists in the tree!")
            
        if root.children:
            for child in root.children:
                self.insert_sibling_by_parent_value_for_all(child, parent, value, is_inserted, is_found, flag=True)
            
        if not flag and not is_found[0]:
            print(f"No match found for parent {parent}") 
        
        return is_inserted[0]

    def insert_sibling_by_grand_parent_and_parent_value(self, root, grand_parent, parent, value, is_inserted=None, is_found=None, flag=False):
        if not root:
            print("General Tree is empty!")
            return False
        
        if is_inserted is None and is_found is None:
            is_inserted = [False]
            is_found = [False]
        
        if root.data == grand_parent:
            is_found[0] = True
            if root.children:
                for child in root.children:
                    if child.data == parent:
                        values = [child.data for child in child.children]
                        if value not in values:
                            is_inserted[0] = True
                            node = GeneralTree(value)
                            child.children.append(node)
                            node.parent = child
                        else:
                            print(f"Value {value} already exists in the tree!")
                            
                        if self.get_root(root, rotation=True) == grand_parent:
                            return is_inserted[0]
                        return (is_inserted, is_found)

            print("Grand parent match found, but not for parent!")
            if self.get_root(root, rotation=True) == grand_parent:
                return is_inserted[0]
            return (is_inserted, is_found)
            
        if root.children:
            for child in root.children:
                result = self.insert_sibling_by_grand_parent_and_parent_value(child, grand_parent, parent, value, is_inserted, is_found, flag=True)
                if isinstance(result, tuple):
                    is_inserted, is_found = result
                if is_inserted[0] or is_found[0]:
                    return is_inserted[0]
            
        if not flag and not is_found[0]:
            print(f"No match found for grand parent {grand_parent}") 
            
        return is_inserted[0]
    
    def insert_sibling_by_grand_parent_and_parent_value_for_all(self, root, grand_parent, parent, value, is_inserted=None, is_found=None, flag=False):
        if not root:
            print("General Tree is empty!")
            return False
        
        if is_inserted is None and is_found is None:
            is_inserted = [False]
            is_found = [False]
        
        if root.data == grand_parent:
            is_found[0] = True
            if root.children:
                for child in root.children:
                    if child.data == parent:
                        values = [child.data for child in child.children]
                        if value not in values:
                            is_inserted[0] = True
                            node = GeneralTree(value)
                            child.children.append(node)
                            node.parent = child
                        else:
                            print(f"Value {value} already exists in the tree!")
                        return

            print("Grand parent match found, but not for parent!")
            
        if root.children:
            for child in root.children:
                self.insert_sibling_by_grand_parent_and_parent_value_for_all(child, grand_parent, parent, value, is_inserted, is_found, flag=True)
            
        if not flag and not is_found[0]:
            print(f"No match found for grand parent {grand_parent}") 
            
        return is_inserted[0]

    def insert_at_under_parent(self, root, parent, index, value, is_inserted=None, is_found=None, flag=False):
        if not root:
            print("General Tree is empty!")
            return False
        
        if is_inserted is None and is_found is None:
            is_inserted = [False]
            is_found = [False]
        
        if root.data == parent:
            is_found[0] = True
            if index <= len(root.children):
                values = [child.data for child in root.children]
                if value not in values:
                    is_inserted[0] = True
                    node = GeneralTree(value)
                    root.children.insert(index, node)
                    node.parent = root
                else:
                    print(f"Value {value} already exists in the tree!")
            else:
                print(f"Parent {parent} matched, but invalid Index!")
                
            if self.get_root(root, rotation=True) == parent:
                return is_inserted[0]
            return (is_inserted, is_found)
        
        if root.children:
            for child in root.children:
                result = self.insert_at_under_parent(child, parent, index, value, is_inserted, is_found, flag=True)
                if isinstance(result, tuple):
                    is_inserted, is_found = result
                if is_inserted[0] or is_found[0]:
                    return is_inserted[0]
                
        if not flag and not is_found[0]:
            print(f"No match found for parent {parent}")
            
        return is_inserted[0]
    
    def insert_at_under_parent_for_all(self, root, parent, index, value, is_inserted=None, is_found=None, flag=False):
        if not root:
            print("General Tree is empty!")
            return False
        
        if is_inserted is None and is_found is None:
            is_inserted = [False]
            is_found = [False]
        
        if root.data == parent:
            is_found[0] = True
            if index <= len(root.children):
                values = [child.data for child in root.children]
                if value not in values:
                    is_inserted[0] = True
                    node = GeneralTree(value)
                    root.children.insert(index, node)
                    node.parent = root
                else:
                    print(f"Value {value} already exists in the tree!")
                return
        
            print(f"Parent {parent} matched, but invalid Index!")
        
        if root.children:
            for child in root.children:
                self.insert_at_under_parent_for_all(child, parent, index, value, is_inserted, is_found, flag=True)
                
        if not flag and not is_found[0]:
            print(f"No match found for parent {parent}")
            
        return is_inserted[0]
    
    def delete_by_value(self, root, value, flag=False):
        if not root:
            print("General Tree is empty!")
            return False
        
        data = None
        if self.get_root(root, rotation=True) != value:
            if root.children:
                for index, child in enumerate(root.children):
                    if child.data == value:
                        return root.children.pop(index).data 
        else:
            print(f"Cannot delete the root node {value}")
            return   
        
        if root.children:
            for child in root.children:
                data = self.delete_by_value(child, value, flag=True)
                if data is not None:
                    return data
                
        if not flag and data is None:
            print(f"No match found for {value}")
    
    def delete_by_value_for_all(self, root, value, data=None, is_found=None, flag=False):
        if not root:
            print("General Tree is empty!")
            return
        
        if data is None:
            data = []
            
        if is_found is None:
            is_found = [False]
        
        if self.get_root(root, rotation=True) != value:
            if root.children:
                for index, child in enumerate(root.children):
                    if child.data == value:
                        is_found[0] = True
                        data.append(root.children.pop(index).data)
        else:
            print(f"Cannot delete the root node {value}")
            return 
        
        if root.children:
            for child in root.children:
                self.delete_by_value_for_all(child, value, data, is_found, flag=True)
                
        if not flag and not is_found[0]:
            print(f"No match found for {value}")
            
        if not data:
            return
                    
        return data
    
    def delete_by_parent_and_value(self, root, parent, value, is_found=None, flag=False):
        if not root:
            print("General Tree is empty!")
            return
        
        if is_found is None:
            is_found = [False]
        
        data = None
        if root.data == parent:
            is_found[0] = True
            if root.children:
                for index, child in enumerate(root.children):
                    if child.data == value:
                        return root.children.pop(index).data
                    
            print(f"Parent {parent} matched, but no children match available for {value}")   
            return data
        
        if root.children:
            for child in root.children:
                data = self.delete_by_parent_and_value(child, parent, value, is_found, flag=True)
                if is_found[0]:
                    return data
                
        if not flag and not is_found[0]:
            print(f"No match found for parent {parent}")
                    
        return data
    
    def delete_by_parent_and_value_for_all(self, root, parent, value, data=None, is_found=None, flag=False):        
        if not root:
            print("General Tree is empty!")
            return
        
        if data is None:
            data = []
            
        if is_found is None:
            is_found = [False]
        
        if root.data == parent:
            is_deleted = False
            is_found[0] = True
            if root.children:
                for index, child in enumerate(root.children):
                    if child.data == value:
                        is_deleted = True
                        data.append(root.children.pop(index).data)
                        
            if not is_deleted:
                print(f"Parent {parent} matched, but no children match available for {value}")   
            
        if root.children:
            for child in root.children:
                self.delete_by_parent_and_value_for_all(child, parent, value, data, is_found, flag=True)
                
        if not flag and not is_found[0]:
            print(f"No match found for parent {parent}")
        
        if not data:
            return
                    
        return data
    
    def delete_at_under_parent(self, root, parent, index, is_found=None, flag=False):
        if not root:
            print("General Tree is empty!")
            return
        
        if is_found is None:
            is_found = [False]
        
        data = None
        if root.data == parent:
            is_found[0] = True
            if root.children:
                if index < len(root.children):
                    return root.children.pop(index).data
                
            print(f"Parent {parent} matched, but invalid Index!")  
            return 
        
        if root.children:
            for child in root.children:
                data = self.delete_at_under_parent(child, parent, index, is_found, flag=True)
                if data or is_found[0]:
                    return data
                
        if not flag and not is_found[0]:
            print(f"No match found for parent {parent}")
                    
        return data
    
    def delete_at_under_all_parent(self, root, parent, index, data=None, is_found=None, flag=False):
        if not root:
            print("General Tree is empty!")
            return []
        
        if data is None and is_found is None:
            data = []
            is_found = [False]
        
        if root.data == parent:
            is_found[0] = True
            value = None
            if root.children:
                if index < len(root.children):
                    value = root.children.pop(index).data
                    data.append(value)
                
            if not value:    
                print(f"Parent {parent} matched, but invalid Index!")  
        
        if root.children:
            for child in root.children:
                self.delete_at_under_all_parent(child, parent, index, data, is_found, flag=True)
                
        if not flag and not is_found[0]:
            print(f"No match found for parent {parent}")
            
        if not data:
            return
                    
        return data
    
def read_file(path):
    val = os.path.split(path)[1]
    try:
        with open(path, 'rt') as f:
            try:
                file = json.load(f)
            except JSONDecodeError as e:
                print(f"The file {val} is empty or contains invalid JSON!")
                print(e)
                file = False
    except FileNotFoundError as e:
        print(f"File {val} not found!")
        print(e)
        file = None
    return file
        
def build_tree(root=None, data=None, tree=None):
    if not root and not tree and not data:
        tree = GeneralTree()
        path = [i for i in Path(os.getcwd()).iterdir() if i.is_dir() and re.match("^files$",os.path.split(i)[1])]
        if path:
            folder_path = path[0]
            file_path = [i for i in Path(folder_path).iterdir() if i.is_file() and os.path.split(i)[1] and re.search(r'.*general_tree_[\w\W]+.json', str(i))]
            if file_path:
                file = read_file(file_path[0])
                if file:
                    data = list(file.keys())[0]
                    root = tree.add_child(root, GeneralTree(data))
                    build_tree(root, file[root.data], tree)
                else:   
                    print("Unable to build the tree!")
                    return (tree,root)
            else:
                print("File not found!")
                return (tree, root)
                 
        else:
            print("Path not found!")
            return (tree, root)
    else:
        if data and isinstance(data, dict):
            for child in data.keys():
                node = GeneralTree(child)
                tree.add_child(root, node)
                build_tree(node, data[child], tree)
        elif data and isinstance(data, list):
            for child in data:
                node = GeneralTree(child)
                tree.add_child(root, node)
                
    return (tree,root)
```
