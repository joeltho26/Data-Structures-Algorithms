```python
import json
from pathlib import Path
import os
import re
from json import JSONDecodeError

class GeneralTree:
    def __init__(self, data: str|None=None):
        self.data = data
        self.children = []
        self.parent = None

    def add_child(self, root, data):      
        if not root:
            return data
        
        root.children.append(data)
        data.parent = root

    def get_level(self):
        level = 0
        itr = self.parent
        while itr:
            itr = itr.parent
            level += 1
        return level

    def print_tree(self, root):
        if not root:
            print("General Tree is empty!")
            return
        
        spaces = " " * root.get_level() * 2
        prefix = spaces + "|--" if root.parent else ""
        print(prefix+root.data)
        if root.children:
            for child in root.children:
                self.print_tree(child)
                
    def depth_of(self, root, data, depth = 0):
        if not root:
            print("General Tree is empty!")
            return 
        
        if root.data == data:
            return depth
        
        if root.children:
            depth += 1
            for child in root.children:
                val = self.depth_of(child, data, depth)
                if isinstance(val, int):
                    return val
    
    def get_depth(self, root):
        depth = 0
        if not root:
            print("General Tree is empty!")
            return 0
        
        if root.children:
            for child in root.children:
                depth = max(self.get_depth(child), depth)
            
        return depth + 1

    def height_of(self, root, data):
        if not root:
            print("General Tree is empty!")
            return
        
        if root.data == data:
            return self.get_height(root) - 1
        
        if root.children:
            for child in root.children:
                val = self.height_of(child, data)
                if isinstance(val, int):
                    return val
                    
    def get_root(self, root):
        if not root:
            print("General Tree is empty!") 
            return
        return root.data
    
    def get_height(self, data):
        return self.get_depth(data)

    def get_diameter(self, root):
        if not root:
            print("General Tree is empty!")
            return 0
        
        left, right = 0, 0
        if root.children:
            for child in root.children:
                node = self.get_depth(child)
                if node > left:
                    right, left = left, node
                elif node > right:
                    right = node

        diameter = 0
        for child in root.children:
            diameter = max(diameter, self.get_diameter(child))
        return max(diameter, left+right+1)

    def get_diameter_of(self, root, data):
        if not root:
            print("General Tree is empty!")
            return 0
        
        if root.data == data:
            return self.get_diameter(root)

        if root.children:
            for child in root.children:
                val = self.get_diameter_of(child, data)
                if val and isinstance(val, int):
                    return val
                
    def count_items(self, root, count=0):
        if not root:
            print("General Tree is empty!")
            return 0
        
        if root.children:
            for child in root.children:
                count = self.count_items(child, count)
        return count + 1
            
    def get_leaf_nodes(self, root, leaves=None):
        if leaves is None:
            leaves = []
            
        if not root:
            print("General Tree is empty!")
            return
        
        if not root.children:
            leaves.append(root.data)
        
        if root.children:
            for child in root.children:
                self.get_leaf_nodes(child, leaves)
        return leaves
    
    def get_leaf_nodes_count(self, root):
        return len(self.get_leaf_nodes(root))
              
    def get_parent(self, root, data, parent=None):
        if parent is None:
            parent = []
            
        if not root:
            print("General Tree is empty!")
            return
        
        if root.data == data:
            if root.parent:
                parent.append(root.parent.data)

        if root.children:
            for child in root.children:
                self.get_parent(child, data, parent)
        return parent
    
    def get_grand_parent(self, root, data, grandparent=None):
        if grandparent is None:
            grandparent = []
            
        if not root:
            print("General Tree is empty!")
            return
        
        if root.data == data:
            if root.parent:
                if root.parent.parent:
                    grandparent.append(root.parent.parent.data)
                
        if root.children:
            for child in root.children:
                self.get_grand_parent(child, data, grandparent)   
                
        return grandparent
    
    def get_children(self, root, data, children=None):
        if children is None:
            children = []
            
        if not root:
            print("General Tree is empty!")
            return
        
        if root.data == data:
            if root.children:
                for child in root.children:
                    children.append(child.data)
                    
        if root.children:
            for child in root.children:
                self.get_children(child, data, children)
                
        return children
    
    def get_children_count(self, root, data):
        return len(self.get_children(root, data))
    
    def get_grand_children(self, root, data, grandchildren=None):
        if grandchildren is None:
            grandchildren = []
            
        if not root:
            print("General Tree is empty!")
            return
        
        if data == root.data:
            if root.children:
                for child in root.children:
                    if child.children:
                        for grandchild in child.children:
                            grandchildren.append(grandchild.data)
                            
        if root.children:
            for child in root.children:
                self.get_grand_children(child, data, grandchildren)
            
        return grandchildren
    
    def get_grand_children_count(self, root, data):
        return len(self.get_grand_children(root, data))
    
    def get_count_key(self, root, data):
        count = 0
        if not root:
            print("General Tree is empty!")
            return 0
        
        if root.data == data:
            return 1
        
        if root.children:
            for child in root.children:
                count += self.get_count_key(child, data)
        return count
        
    def search(self, root, data):
        if not root:
            print("General Tree is empty!")
            return False
        
        if root.data == data:
            return True
        
        if root.children:
            for child in root.children:
                res = self.search(child, data)    
                if res:
                    return True
        return False
    
    def get_first_child(self, root, first_children=None):
        if first_children is None:
            first_children = []
        
        if not root:
            print("General Tree is empty!")
            return 
        
        if root.children:
            first_children.append(root.children[0].data)
                
        if root.children:
            for child in root.children:
                self.get_first_child(child, first_children)
                
        return first_children
    
    def get_first_child_of(self, root, data, first_children=None):
        if first_children is None:
            first_children = []
            
        if not root:
            print("General Tree is empty!")
            return 
        
        if root.data == data:
            if root.children:
                first_children.append(root.children[0].data)
                
        if root.children:
            for child in root.children:
                self.get_first_child_of(child, data, first_children)
        return first_children

    def get_middle_child(self, root, middle_children=None):
        if middle_children is None:
            middle_children = []
        
        if not root:
            print("General Tree is empty!")
            return 
        
        if root.children:
            mid_index = len(root.children)//2
            middle_children.append(root.children[mid_index].data)
                
        if root.children:
            for child in root.children:
                self.get_middle_child(child, middle_children)
                
        return middle_children
    
    def get_middle_child_of(self, root, data, middle_children=None): 
        if middle_children is None:
            middle_children = []
                   
        if not root:
            print("General Tree is empty!")
            return
        
        if root.data == data:
            if root.children:
                mid_index = len(root.children)//2
                middle_children.append(root.children[mid_index].data)
                
        if root.children:
            for child in root.children:
                self.get_middle_child_of(child, data, middle_children)
        return middle_children

    def get_last_child(self, root, last_children=None):
        if last_children is None:
            last_children = []
        
        if not root:
            print("General Tree is empty!")
            return
        
        if root.children:
            last_children.append(root.children[-1].data)
                
        if root.children:
            for child in root.children:
                self.get_last_child(child, last_children)
                
        return last_children
        
    def get_last_child_of(self, root, data, last_children=None):
        if last_children is None:
            last_children = []
            
        if not root:
            print("General Tree is empty!")
            return
        
        if root.data == data:
            if root.children:
                last_children.append(root.children[-1].data)
                
        if root.children:
            for child in root.children:
                self.get_last_child_of(child, data, last_children)
        return last_children
    
    def dfs_traversal(self, root, data=None):
        if data is None:
            data = []

        if not root:
            print("General Tree is empty!")
            return
        
        data.append(root.data)
        
        if root.children:
            for child in root.children:
                self.dfs_traversal(child, data)
                
        return data
    
    def bfs_traversal(self, root, data=None):
        if data is None:
            data = []
            
        if not root:
            print("General Tree is empty!")
            return
        
        q = [root]
        while q:
            for _ in range(len(q)):
                node = q.pop(0)
                data.append(node.data)
                
                if node.children:
                    for child in node.children:
                        q.append(child)
        return data
    
    def insert_by_value(self, root, value, data):
        if not root:
            print("General Tree is empty!")
            return
        
        if root.data == value:
            if root.parent:
                node = GeneralTree(data)
                self.add_child(root.parent, node)
                print(f"Inserted node {data} as sibling to Cell Phone")
                return True
            
            print(f"No parent available for {value} to insert!")
            return
                
        if root.children:
            for child in root.children:
                res = self.insert_by_value(child, value, data)
                if res:
                    return 
                
    def insert_by_grand_parent_and_parent_value(self, root, grandparent, parent, data):
        if not root:
            print("General Tree is empty!")
            return
        
        parent_not_found = True
        if root.data == grandparent:
            if root.children:
                for child in root.children:
                    if child.data == parent:
                        self.add_child(child, GeneralTree(data))
                        parent_not_found = False
                        print(f"Inserted node {data} under grandparent {grandparent} and parent {parent}")
                        return True
                if parent_not_found:
                    print(f"No matching child for grandparent {grandparent} with parent value {parent}")
                    return
            print(f"No child available for grandparent {grandparent}")
            return
                    
        if root.children:
            for child in root.children:
                res = self.insert_by_grand_parent_and_parent_value(child, grandparent, parent, data)
                if res:
                    return
        
    def insert_at_under_parent(self, root, parent, index, data):
        if not root:
            print("General Tree is empty!")
            return
        
        if root.data == parent:
            if root.children:
                if index <= len(root.children):
                    node = GeneralTree(data)
                    node.parent = root
                    root.children.insert(index, node)
                    print(f"Inserted node {data} at index {index} under parent {parent}")
                    return True
                else:
                    raise Exception("Invalid Index")
            print(f"No children found for parent {parent}")
            return False
                
        if root.children:
            for child in root.children:
                res = self.insert_at_under_parent(child, parent, index, data)
                if res:
                    return
                
    def delete_by_value(self, root, value, found=False):
        if not root:
            print("General Tree is empty!")
            return
        
        if not found:
            found = self.search(root, value)
            if not found:
                print("Value not found in Tree to delete!")
                return
        
        if root.data == value:
            if root.parent is None:
                print(f"Cannot delete the root node {value}!")
                return
            else:
                root.parent.children.remove(root)
                print(f"Deleted node with value {value}")
                return True
            
        if root.children:
            for child in root.children:
                res = self.delete_by_value(child, value, found)
                if res:
                    return
    
    def delete_by_parent_and_value(self, root, parent, value):
        if not root:
            print("General Tree is empty!")
            return
        
        parent_not_found = True
        if root.data == parent:
            if root.children:
                for index, child in enumerate(root.children):
                    if child.data == value:
                        root.children.pop(index)
                        parent_not_found = False
                        print(f"Deleted node with value {value} under parent {parent}")
                        return value
                if parent_not_found:
                    print(f"No matching child for parent {parent}")
                    return
            print(f"No children found for parent {parent}")
            
        if root.children:
            for child in root.children:
                res = self.delete_by_parent_and_value(child, parent, value)
                if res:
                    return value
                
    def delete_at_under_parent(self, root, parent, index):
        if not root:
            print("General Tree is empty!")
            return
        
        if root.data == parent:
            if root.children:
                if index < len(root.children):
                    val = root.children.pop(index)
                    print(f"Deleted node {val.data} at index {index} under parent {parent}")
                    return val.data
                else:
                    raise Exception("Invalid Index")
            print(f"No children found for parent {parent}")
            return
                
        if root.children:
            for child in root.children:
                res = self.delete_at_under_parent(child, parent, index)
                if res:
                    return res
    
def read_file(path):
    val = os.path.split(path)[1]
    try:
        with open(path, 'rt') as f:
            try:
                file = json.load(f)
            except JSONDecodeError as e:
                print(f"The file {val} is empty or contains invalid JSON!")
                print(e)
                file = False
    except FileNotFoundError as e:
        print(f"File {val} not found!")
        print(e)
        file = None
    return file
        
def build_tree(root=None, data=None, tree=None):
    if not root and not tree and not data:
        tree = GeneralTree()
        path = [i for i in Path(os.getcwd()).iterdir() if i.is_dir() and re.match("^files$",os.path.split(i)[1])]
        if path:
            folder_path = path[0]
            file_path = [i for i in Path(folder_path).iterdir() if i.is_file() and os.path.split(i)[1]]
            if file_path:
                file = read_file(file_path[0])
                if file:
                    data = list(file.keys())[0]
                    root = tree.add_child(root, GeneralTree(data))
                    build_tree(root, file[root.data], tree)
                else:   
                    print("Unable to build the tree!")
                    return (tree,root)
            else:
                print("File not found!")
                return (tree, root)
                 
        else:
            print("Path not found!")
            return (tree, root)
    else:
        if data and isinstance(data, dict):
            for child in data.keys():
                node = GeneralTree(child)
                tree.add_child(root, node)
                build_tree(node, data[child], tree)
        elif data and isinstance(data, list):
            for child in data:
                node = GeneralTree(child)
                tree.add_child(root, node)
                
    return (tree,root)
```
