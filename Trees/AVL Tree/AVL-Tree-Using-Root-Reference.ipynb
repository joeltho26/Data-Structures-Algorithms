```python
from typing import Optional

class AVLTree:
    def __init__(self, data: int|None=None):
        self.data = data
        self.left = None
        self.right = None
        self.height = 1
        self.is_root = False
        
    def add_child(self, root, val):
        if not root:
            return val
        elif val.data < root.data:
            root.left = self.add_child(root.left, val)
        elif val.data > root.data:
            root.right = self.add_child(root.right, val)
        
        balance = self.get_balance(root)
        root.height = 1 + max(self.get_height(root.left, True), self.get_height(root.right, True))
        
        if balance > 1 and val.data < root.left.data:
            return self.rotate_right(root)
        if balance < -1 and val.data > root.right.data:
            return self.rotate_left(root)
        if balance > 1 and val.data > root.left.data:
            root.left = self.rotate_left(root.left)
            return self.rotate_right(root)
        if balance < -1 and val.data < root.right.data:
            root.right = self.rotate_right(root.right)
            return self.rotate_left(root)
        
        return root

    def search(self, root, data, validate: Optional[bool]=False, flag=False):
        if not root:
            print("The AVL Tree is empty!")
            return False
        
        if root.data == data:
            if not validate:
                print(f"{data} found in the AVL Tree!")
            return True
        
        if data < root.data:
            if root.left:
                return self.search(root.left, data, flag=True)
        elif data > root.data:
            if root.right:
                return self.search(root.right, data, flag=True)
        
        if not flag:
            if not validate:
                print(f"{data} not found in the AVL Tree!")
            return False
    
    def inorder_traversal(self, root):
        if not root:
            print("The AVL Tree is empty!")
            return
        
        elements = []
        
        if root.left:
            elements += self.inorder_traversal(root.left)
            
        elements.append(root.data)
        
        if root.right:
            elements += self.inorder_traversal(root.right)
            
        return elements

    def preorder_traversal(self, root):
        if not root:
            print("The AVL Tree is empty!")
            return
        
        elements = []
        
        elements.append(root.data)
        
        if root.left:
            elements += self.preorder_traversal(root.left)
        
        if root.right:
            elements += self.preorder_traversal(root.right)
            
        return elements
    
    def postorder_traversal(self, root):
        if not root:
            print("The AVL Tree is empty!")
            return
        
        elements = []
        
        if root.left:
            elements += self.postorder_traversal(root.left)
        
        if root.right:
            elements += self.postorder_traversal(root.right)
            
        elements.append(root.data)
            
        return elements
    
    def get_count_of_items(self, root):
        count = 0
        if root.left:
            count += self.get_count_of_items(root.left)
    
        if root.right:
            count += self.get_count_of_items(root.right)
            
        return count + 1
    
    def get_children(self, root, data, is_found=None, flag=False):
        if not root:
            print("The AVL tree is empty!")
            return
        
        if is_found is None:
            is_found = [False]
        
        if data == root.data:
            is_found[0] = True
            children = []
            if root.left:
                children.append(root.left.data)
            if root.right:
                children.append(root.right.data)
            
            if not children:
                print(f"Match found, but children not available for {data}")
            return children
                
        if data < root.data: 
            if root.left:
                val = self.get_children(root.left, data, is_found, flag=True)
                if isinstance(val, list):
                    if not val:
                        return
                    return val
        elif data > root.data:
            if root.right:
                val = self.get_children(root.right, data, is_found, flag=True)
                if isinstance(val, list):
                    if not val:
                        return
                    return val
                
        if not flag and not is_found[0]:
            print(f"{data} not found in the AVL Tree!")

    def get_balance(self, root):
        if not root:
            return 0
        
        return self.get_height(root.left, True) - self.get_height(root.right, True)
    
    def get_height(self, root, other: bool|None=False):
        if not root:
            if not other:
                print("The AVL Tree is empty!")
            return 0
        return root.height
    
    def depth(self, root):
        return self.get_height(root)
    
    def get_diameter(self, root):
        ldepth, rdepth = 0, 0
        ldiameter, rdiameter = 0, 0
        if root.left:
            ldepth = self.depth(root.left)
            ldiameter = self.get_diameter(root.left)
            
        if root.right:
            rdepth = self.depth(root.right)
            rdiameter = self.get_diameter(root.right)
            
        return max(ldepth+rdepth+1, max(ldiameter, rdiameter))
    
    def diameter_of(self, root, data, flag=False):
        if not root:
            print("The AVL Tree is empty!")
            return 
        
        diameter = None
        if data == root.data:
            return self.get_diameter(root)
        
        if data < root.data:
            if root.left:
                diameter = self.diameter_of(root.left, data, flag=True)
                if diameter:
                    return diameter
                
        elif data > root.data:
            if root.right:
                diameter = self.diameter_of(root.right, data, flag=True)
                if diameter:
                    return diameter
            
        if not flag and diameter is None:
            print(f"{data} not found in the AVL Tree!")
    
    def height_of(self, root, data, is_found=None, flag=False):
        if not root:
            print("The AVL Tree is empty!")
            return
        
        if is_found is None:
            is_found = [False]
        
        height = None
        if root.data == data:
            is_found[0] = True
            return root.height - 1
        
        if data < root.data:
            if root.left:
                height = self.height_of(root.left, data, is_found, flag=True)
                if height or is_found[0]:
                    return height
        elif data > root.data:
            if root.right:
                height = self.height_of(root.right, data, is_found, flag=True) 
                if height or is_found[0]:
                    return height
            
        if not flag and not is_found[0]:
            print(f"{data} not found in the AVL Tree!")
            
    
    def depth_of(self, root, data, depth=0, flag=False):
        if not root:
            print("The AVL Tree is empty!")
            return
        
        if root.data == data:
            return depth
        
        if data < root.data:
            if root.left:
                depth += 1
                depth = self.depth_of(root.left, data, depth, flag=True)
                if depth:
                    return depth
                
        elif data > root.data:
            if root.right:
                depth += 1
                depth = self.depth_of(root.right, data, depth, flag=True)
                if depth:
                    return depth
            
        if not flag and not depth:
            print(f"{data} not found in the AVL Tree!")
        
    
    def rotate_right(self, z):
        y = z.left
        T2 = y.right
        
        # rotation
        y.right = z
        z.left = T2

        z.height = 1 + max(self.get_height(z.left, True), self.get_height(z.right, True))
        y.height = 1 + max(self.get_height(y.left, True), self.get_height(y.right, True))
        
        return y
    
    def rotate_left(self, z):
        y = z.right
        T3 = y.left
        
        # rotation
        y.left = z
        z.right = T3

        z.height = 1 + max(self.get_height(z.left, True), self.get_height(z.right, True))
        y.height = 1 + max(self.get_height(y.left, True), self.get_height(y.right, True))
        
        return y
    
    def calculate_sum(self, root):
        if not root:
            print("The AVL Tree is empty!")
            return 0
        
        sum = 0
        if root.left:
            sum += self.calculate_sum(root.left)
            
        if root.right:
            sum += self.calculate_sum(root.right)
            
        return sum + root.data
    
    def get_left_sum(self, root):
        if not root:
            print("The AVL Tree is empty!")
            return 0
        
        sum = 0
        if root.left:
            sum += self.calculate_sum(root.left)
            
        return sum
    
    def get_right_sum(self, root):
        if not root:
            print("The AVL Tree is empty!")
            return 0
        
        sum = 0
        if root.right:
            sum += self.calculate_sum(root.right)
            
        return sum
    
    def get_root(self, root):
        if not root:
            print("The AVL Tree is empty!")
            return 
        
        return root.data
    
    def get_min_value(self, root):
        if not root:
            print("The AVL Tree is empty!")
            return 
        
        if root.left:
            return self.get_min_value(root.left)
            
        return root.data
    
    def get_max_value(self, root):
        if not root:
            print("The AVL Tree is empty!")
            return 
        
        if root.right:
            return self.get_max_value(root.right)
            
        return root.data
    
    def dfs_traversal(self, root):
        return self.preorder_traversal(root)
    
    def bfs_traversal(self, root):
        if not root:
            print("The AVL Tree is empty!")
            return
        
        q = [root]
        res = []
        while q:
            for _ in range(len(q)):
                node = q.pop(0)
                res.append(node.data)
                
                if node.left:
                    q.append(node.left)
                    
                if node.right:
                    q.append(node.right)
                    
        return res
    
    def get_parent(self, root, data, is_found=None, flag=False):
        if not root:
            print("AVL Tree not available")
            return
        
        if is_found is None:
            is_found = [False]
            
        if root.data == data and root.is_root:
            is_found[0] = True
            print(f"Match found, {data} is the root node!")
            return
        
        if root.left:
            if root.left.data == data:
                is_found[0] = True
                return root.data
                
        if root.right:
            if root.right.data == data:
                is_found[0] = True
                return root.data
            
        if data < root.data:
            if root.left:
                parent = self.get_parent(root.left, data, is_found, flag=True)
                if parent or is_found[0]:
                    if not parent:
                        return
                    return parent
                    
        elif data > root.data:
            if root.right:
                parent = self.get_parent(root.right, data, is_found, flag=True)
                if parent or is_found[0]:
                    if not parent:
                        return
                    return parent
                
        if not flag and not is_found[0]:
            print(f"{data} not found in the AVL Tree!")
            
    def get_grand_children(self, root, data, is_children=None, is_found=None, flag=False):
        if not root:
            print("AVL Tree not available")
            return
        
        if is_children is None and is_found is None:
            is_children = [False]
            is_found = [False]
        
        grandchildren = None
        if root.data == data:
            is_found[0] = True
            grandchildren = []
            if root.left:
                is_children[0] = True
                if root.left.left:
                    grandchildren.append(root.left.left.data)
                if root.left.right:
                    grandchildren.append(root.left.right.data)
            if root.right:
                is_children[0] = True
                if root.right.left:
                    grandchildren.append(root.right.left.data)
                if root.right.right:
                    grandchildren.append(root.right.right.data)
                    
            if not grandchildren:
                if not is_children[0]:
                    print(f"No children and grandchildren found for {data}")
                    return grandchildren
                
                print(f"No grandchildren found for {data}")
                return grandchildren

            return grandchildren
            
        if data < root.data:
            if root.left:
                grandchildren = self.get_grand_children(root.left, data, is_children, is_found, flag=True)
                if isinstance(grandchildren, list):
                    if not grandchildren:
                        return
                    return grandchildren
        elif data > root.data:
            if root.right:
                grandchildren = self.get_grand_children(root.right, data, is_children, is_found, flag=True)
                if isinstance(grandchildren, list):
                    if not grandchildren:
                        return
                    return grandchildren
                
        if not flag and not is_found[0]:
            print(f"{data} not found in the AVL Tree!")
    
    def get_grand_parent(self, root, data, is_found=None, flag=False):
        if not root:
            print("AVL Tree not available")
            return
        
        if is_found is None:
            is_found = [False]
        
        if root.data == data and root.is_root:
            is_found[0] = True
            print(f"No grandparent and parent found for {data} because its the root node!")
            return
        
        if root.left:
            if root.left.left:
                if root.left.left.data == data:
                    is_found[0] = True
                    return root.data                        
                    
            if root.left.right:
                if root.left.right.data == data:
                    is_found[0] = True
                    return root.data
            
            if root.left.data == data and root.is_root:
                is_found[0] = True
                print(f"No grand parent found for {data}")
                return
                
        if root.right:
            if root.right.right:
                if root.right.right.data == data:
                    is_found[0] = True
                    return root.data
                
            if root.right.left:
                if root.right.left.data == data:
                    is_found[0] = True
                    return root.data                    
                
            if root.right.data == data and root.is_root:
                is_found[0] = True
                print(f"No grand parent found for {data}")
                return
            
        if data < root.data:
            if root.left:
                val = self.get_grand_parent(root.left, data, is_found, flag=True)
                if val or is_found[0]:
                    if not val:
                        return
                    return val

        elif data > root.data:
            if root.right:
                val = self.get_grand_parent(root.right, data, is_found, flag=True)
                if val or is_found[0]:
                    if not val:
                        return
                    return val
            
        if not flag and not is_found[0]:
            print(f"{data} not found in the AVL Tree!")

    
    def get_leaf_nodes(self, root, leaves=None):
        if leaves is None:
            leaves = []
            
        if not root:
            print("AVL Tree not available")
            return
        
        if root.left:
            if not root.left.left and not root.left.right:
                leaves.append(root.left.data)
            self.get_leaf_nodes(root.left, leaves)

        if root.right:
            if not root.right.left and not root.right.right:
                leaves.append(root.right.data)
            self.get_leaf_nodes(root.right, leaves)

        return leaves
    
    def delete(self, root, data, is_found=None, flag=False):
        if not root:
            print("AVL Tree not available")
            return 
        
        if is_found is None:
            is_found = [False]
            
        if data < root.data:
            if root.left:
                root.left = self.delete(root.left, data, is_found, flag=True)
        elif data > root.data:
            if root.right:
                root.right = self.delete(root.right, data, is_found, flag=True)
        else:
            is_found[0] = True
            if root.left is None and root.right is None:
                return None
            elif root.right is None:
                tmp = root.left
                root = None
                return tmp
            elif root.left is None:
                tmp = root.right
                root = None
                return tmp
            
            # right rotation
            min_val = self.get_min_value(root.right)
            root.data = min_val
            root.right = self.delete(root.right, min_val, is_found, flag=True)
            
            # right rotation
            # max_val = self.get_min_value(root.left)
            # root.data = max_val
            # root.left = self.delete(root.left, max_val)
        if not root:
            return None
        
        balance = self.get_balance(root)
        root.height = 1 + max(self.get_height(root.left, True), self.get_height(root.right, True))
        
        if balance > 1 and self.get_balance(root.left) >= 0:
            return self.rotate_right(root)
        if balance > 1 and self.get_balance(root.left) < 0:
            root.left = self.rotate_left(root.left)
            return self.rotate_right(root)
        if balance < -1 and self.get_balance(root.right) <= 0:
            return self.rotate_left(root)
        if balance < -1 and self.get_balance(root.right) > 0:
            root.right = self.rotate_right(root.right)
            return self.rotate_left(root)
        
        if not flag and not is_found[0]:
            print(f"{data} not found in the AVL Tree!")
        
        return root
    
def build_tree(numbers, root=None, tree=None):
    tree = AVLTree()
    if numbers:
        if root is None:
            root = tree.add_child(root, AVLTree(numbers[0]))
            root.is_root = True
        for i in numbers[1:]:
            root = tree.add_child(root, AVLTree(i))
        return (tree, root)    
    else:
        print("Unable to create AVL Tree!")
        return (tree, root)
    
    
```
