```python
from typing import Optional

class BinarySearchTree:
    def __init__(self, data: int|None=None):
        self.data = data
        self.right = None
        self.left = None
        self.is_root = False

    def add_child(self, root, value):
        if not root:
            return value
        elif value.data < root.data:
            root.left = self.add_child(root.left, value)
        elif value.data > root.data:
            root.right = self.add_child(root.right, value)
        return root
    
    def search(self, root, data, validate: Optional[bool]=False, flag=False):
        if not root:
            print("Binary Search Tree is empty!")
            return False
        
        if root.data == data:
            if not validate:
                print(f"{data} is present in the Binary Search Tree!")
            return True

        if data < root.data:
            if root.left:
                val = self.search(root.left, data, flag=True)
                if val:
                    return val
        elif data > root.data:
            if root.right:
                val = self.search(root.right, data, flag=True)
                if val:
                    return val
            
        if not flag:
            if not validate:
                print(f"{data} is not present in the Binary Search Tree!")
            return False
    
    def get_count_of_items(self, root):
        if not root:
            print("Binary Search Tree is empty!")
            return 0
        
        count = 0
        if root.left:
            count += self.get_count_of_items(root.left)
        
        if root.right:
            count += self.get_count_of_items(root.right)
        
        return count + 1
    
    def inorder_traversal(self, root):
        if not root:
            print("Binary Search Tree is empty!")
            return
        
        elements = []
        if root.left:
            elements += self.inorder_traversal(root.left)
            
        elements.append(root.data)
            
        if root.right:
            elements += self.inorder_traversal(root.right)
            
        return elements 
    
    def preorder_traversal(self, root):
        if not root:
            print("Binary Search Tree is empty!")
            return
        
        elements = []
        
        elements.append(root.data)
        
        if root.left:
            elements += self.preorder_traversal(root.left)
            
        if root.right:
            elements += self.preorder_traversal(root.right)
            
        return elements 
    
    def postorder_traversal(self, root):
        if not root:
            print("Binary Search Tree is empty!")
            return
        
        elements = []
        if root.left:
            elements += self.postorder_traversal(root.left)
            
        if root.right:
            elements += self.postorder_traversal(root.right)
            
        elements.append(root.data)
            
        return elements 
    
    def get_children(self, root, data, is_found=None, flag=False):
    
        children = None 
        if not root:
            print("Binary Search Tree is empty!")
            return
        
        if is_found is None:
            is_found = [False]
        
        if data == root.data:
            children = []
            is_found[0] = True
            if root.left:
                children.append(root.left.data)
            if root.right:
                children.append(root.right.data)
            
            if not children:
                print("Match found, but children not available for {data}")
            
            return children
                
        if data < root.data:
            if root.left:
                val = self.get_children(root.left, data, is_found, flag=True)
                if isinstance(val, list):
                    if not val:
                        return
                    return val
                    
        if data > root.data:
            if root.right:
                val = self.get_children(root.right, data, children, flag=True)
                if isinstance(val, list):
                    if not val:
                        return
                    return val
                    
        if not flag and not is_found[0]:
            print(f"{data} not found in Binary Search tree!")
            return
    
    def calculate_sum(self, root):
        if not root:
            print("Binary Search Tree is empty!")
            return 0 
        
        sum = 0
        if root.left:
            sum += self.calculate_sum(root.left)
            
        if root.right:
            sum += self.calculate_sum(root.right)
            
        return sum + root.data
    
    def get_left_sum(self, root):
        if not root:
            print("Binary Search Tree is empty!")
            return 0 
        
        if root.left:
            return self.calculate_sum(root.left)
    
    def get_right_sum(self, root):
        if not root:
            print("Binary Search Tree is empty!")
            return 0 
        
        if root.right:
            return self.calculate_sum(root.right)
        
    def get_root(self, root):
        if not root:
            print("Binary Search Tree is empty!")
            return
            
        if root.is_root:
            return root.data
        
    def get_min_value(self, root):
        if not root:
            print("Binary Search Tree is empty!")
            return
        
        if root.left:
            return self.get_min_value(root.left)
            
        return root.data
    
    def get_max_value(self, root):
        if not root:
            print("Binary Search Tree is empty!")
            return
        
        if root.right:
            return self.get_max_value(root.right)
        
        return root.data
        
    def get_height(self, root):
        if not root:
            print("Binary Search Tree is empty!")
            return 0
        
        lheight, rheight = 0, 0
        if root.left:
            lheight = self.get_height(root.left)
            
        if root.right:
            rheight = self.get_height(root.right)
    
        return max(lheight,rheight) + 1
    
    def get_depth(self, root):
        return self.get_height(root)
        
    def height_of(self, root, data, flag=False):
        if not root:
            print("Binary Search Tree is empty!")
            return 
        
        height = None
        if root.data == data:
            return self.get_height(root) - 1
        
        if data < root.data:
            if root.left:
                height = self.height_of(root.left, data, flag=True)
                if height is not None:
                    return height
        elif data > root.data:
            if root.right:
                height = self.height_of(root.right, data, flag=True)
                if height is not None:
                    return height
                
        if not flag:
            print(f"The value {data} is not present in the Binary Search Tree!")   

    def depth_of(self, root, data, depth=0, flag=False):
        if not root:
            print("Binary Search Tree is empty!")
            return
        
        if root.data == data:
            return depth
        
        if data < root.data:
            if root.left:
                depth += 1
                depth = self.depth_of(root.left, data, depth, flag=True)
                if depth:
                    return depth
                
        elif data > root.data:
            if root.right:
                depth += 1
                depth = self.depth_of(root.right, data, depth, flag=True)
                if depth:
                    return depth
                
        if not flag and not depth:
            print(f"The value {data} is not present in the Binary Search Tree!") 

    def get_diameter(self, root):
        if not root:
            print("Binary Search Tree is empty!")
            return 0
        
        ldepth, rdepth = 0, 0
        ldiameter, rdiameter = 0, 0
        if root.left:
            ldepth = self.get_depth(root.left)
            ldiameter = self.get_diameter(root.left)

        if root.right:
            rdepth = self.get_depth(root.right)
            rdiameter = self.get_diameter(root.right)
            
        return max(max(ldiameter,rdiameter), ldepth+rdepth+1)

    def diameter_of(self, root, data, flag=False):
        diameter = None
        if not root:
            print("Binary Search Tree is empty!")
            return
        
        if root.data == data:
            return self.get_diameter(root)

        if data < root.data:
            if root.left:
                diameter = self.diameter_of(root.left, data, flag=True)
                if diameter:
                    return diameter
        elif data > root.data:
            if root.right:
                diameter = self.diameter_of(root.right, data, flag=True)
                if diameter:
                    return diameter
                
        if not flag and diameter is None:
            print(f"The value {data} is not present in the Binary Search Tree!")

    def dfs_traversal(self, root):
        return self.preorder_traversal(root)

    def bfs_traversal(self, root):
        res = []
        if not root:
            print("Binary Search Tree is empty!")
            return res
        
        q = [root]
        while q:
            for _ in range(len(q)):
                node = q.pop(0)
                res.append(node.data)
                
                if node.left:
                    q.append(node.left)
                    
                if node.right:
                    q.append(node.right)
                    
        return res
    
    def get_parent(self, root, data, is_found=None, flag=False):
        if not root:
            print("Binary Search Tree is empty!")
            return 
        
        if is_found is None:
            is_found = [False]
            
        if root.data == data and root.is_root:
            is_found[0] = True
            print(f"Match found is a root node {data} and there are no parent available for it!")
            return
        
        if root.left:
            if root.left.data == data:
                is_found[0] = True
                return root.data
        if root.right:
            if root.right.data == data:
                is_found[0] = True
                return root.data
            
        if data < root.data:
            if root.left:
                parent = self.get_parent(root.left, data, is_found, flag=True)
                if parent or is_found[0]:
                    if not parent:
                        return
                    return parent
                    
        elif data > root.data:
            if root.right:
                parent = self.get_parent(root.right, data, is_found, flag=True)
                if parent or is_found[0]:
                    if not parent:
                        return
                    return parent
            
        if not flag and not is_found[0]:
            print(f"{data} not found in the Binary Search Tree!")
                

    def get_grand_children(self, root, data, is_found=None, flag=False):
        if not root:
            print("Binary Search Tree is empty!")
            return 
        
        if is_found is None:
            is_found = [False]
        
        if root.data == data:
            is_found[0] = True
            grandchildren = []
            if root.left:
                if root.left.left:
                    grandchildren.append(root.left.left.data)
                if root.left.right:
                    grandchildren.append(root.left.right.data)
            if root.right:
                if root.right.left:
                    grandchildren.append(root.right.left.data)
                if root.right.right:
                    grandchildren.append(root.right.right.data)
            return grandchildren
        
        if data < root.data:
            if root.left:
                grandchildren = self.get_grand_children(root.left, data, flag=True)
                if grandchildren or is_found[0]:
                    if not grandchildren:
                        return
                    return grandchildren
        elif data > root.data:
            if root.right:
                grandchildren = self.get_grand_children(root.right, data, flag=True)
                if grandchildren or is_found[0]:
                    if not grandchildren:
                        return
                    return grandchildren
            
        if not flag and not is_found[0]:
            print(f"{data} not found in the Binary Search Tree!")
            
    def get_grand_parent(self, root, data, is_found=None, flag=False):
        if not root:
            print("Binary Search Tree is empty!")
            return 
        
        if is_found is None:
            is_found = [False]
        
        if root.data == data and root.is_root:
            is_found[0] = True
            print(f"Match found, {data} is the root node of Binary Search Tree and has not parent!")
            return
            
        if root.left and root.is_root:
            if root.left.data == data:
                is_found[0] = True
                print(f"Match found, {data} but has no grandparent!")
                return

        if root.right and root.is_root:
            if root.right.data == data:
                is_found[0] = True
                print(f"Match found, {data} but has no grandparent!")
                return
        
        if root.left:
            if root.left.left:
                if root.left.left.data == data:
                    is_found[0] = True
                    return root.data
            if root.left.right:
                if root.left.right.data == data:
                    is_found[0] = True
                    return root.data
        if root.right:
            if root.right.left:
                if root.right.left.data == data:
                    is_found[0] = True
                    return root.data

            if root.right.right:
                if root.right.right.data == data:
                    is_found[0] = True
                    return root.data
            
        if data < root.data:
            if root.left:
                grandparent = self.get_grand_parent(root.left, data, is_found, flag=True)
                if grandparent or is_found[0]:
                    if not grandparent:
                        return
                    return grandparent
        elif data > root.data:
            if root.right:
                grandparent = self.get_grand_parent(root.right, data, is_found, flag=True)
                if grandparent or is_found[0]:
                    if not grandparent:
                        return
                    return grandparent
                
        if not flag and not is_found[0]:
            print(f"{data} not found in the Binary Search Tree!")
            
    def get_leaf_nodes(self, root, leaves=None):
        if leaves is None:
            leaves = []
            
        if not root:
            print("Binary Search Tree is empty!")
            return 
        
        if not root.left and not root.right:
            leaves.append(root.data)
            
        if root.left:
            self.get_leaf_nodes(root.left, leaves)
            
        if root.right:
            self.get_leaf_nodes(root.right, leaves)
        
        return leaves

    def delete(self, root, data, is_found=None, flag=False):
        if not root:
            print("Binary Search Tree is empty!")
            return 
        
        if is_found is None:
            is_found = [False]
        
        if data < root.data:
            if root.left:
                root.left = self.delete(root.left, data, is_found, flag=True)
        elif data > root.data:
            if root.right:
                root.right = self.delete(root.right, data, is_found, flag=True)
        else:
            is_found[0] = True
            if not root.left and not root.right:
                return None
            elif not root.left:
                return root.right
            elif not root.right:
                return root.left
            
            # right rotation
            min_val = self.get_min_value(root.right)
            root.data = min_val
            root.right = self.delete(root.right, min_val, is_found, flag=True)
            
            # left rotation
            # max_val = self.get_max_value(root.left)
            # root.data = max_val
            # root.left = self.delete(root.left, max_val)
            
        if not flag and not is_found[0]:
            print(f"{data} not found in the Binary Search Tree!")
                
        return root 

def build_tree(numbers, root=None, tree=None):
    if numbers:
        if root is None and tree is None:
            tree = BinarySearchTree()
            root = tree.add_child(root, BinarySearchTree(numbers[0]))
            root.is_root = True
        for i in numbers[1:]:
            root = tree.add_child(root, BinarySearchTree(i))
    else:
        print("No values available to build the tree!")
        return
    
    return (tree, root)
```
