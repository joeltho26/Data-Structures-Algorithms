```python
# Binary Search Tree
from typing import Optional

class BST:
    def __init__(self, data=None):
        self.data = data
        self.left = None
        self.right = None
        self.parent = False
        
    def add_child(self, data):
        if not self:
            return BST(data)
        
        if data < self.data:
            if self.left:
                self.left.add_child(data)
            else:
                self.left = BST(data)
        elif data > self.data:
            if self.right:
                self.right.add_child(data)
            else:
                self.right = BST(data)
                
    def search(self, value, validate: Optional[bool]=False, flag=False):
        if not self:
            print("The Binary Search Tree is empty!")
            return False
        
        if self.data == value:
            if not validate:
                print(f"Match found for {value}")
            return True
        
        if value < self.data:
            if self.left:
                return self.left.search(value)
        elif value > self.data:
            if self.right:
                return self.right.search(value) 
        
        if not flag:
            if not validate:
                print(f"No match found for {value}")
            return False

    def get_root(self, validate: Optional[bool]=False):
        if not self:
            print("The Binary Search Tree is empty!")
            return 
        
        if self.parent:
            if not validate:
                print(f"The root of the Binary Search Tree is: {self.data}")
            return self.data
    
    def get_count_of_items(self):
        if not self:
            print("The Binary Search Tree is empty!")
            return 
            
        count = 1
        if self.left:
            count += self.left.get_count_of_items()
            
        if self.right:
            count += self.right.get_count_of_items()
            
        return count

    def delete_child(self, value, is_found=None, flag=False):
        if not self:
            print("The Binary Search Tree is empty!")
            return
        
        if is_found is None:
            is_found = [False]
        
        if value < self.data:
            if self.left:
                self.left = self.left.delete_child(value, is_found, flag=True)
        elif value > self.data:
            if self.right:
                self.right = self.right.delete_child(value, is_found, flag=True)
        else:
            is_found[0] = True
            if self.left is None and self.right is None:
                return None
            elif self.left is None:
                return self.right      
            elif self.right is None:
                return self.left
            
            # taking right node to replace the deleted node
            min_value = self.right.get_min_value()
            self.data = min_value
            self.right = self.right.delete_child(min_value)
            
            # taking left node to replace the deleted node
            # max_value = self.left.get_max_value()
            # self.data = max_value
            # self.left = self.left.delete_child(max_value, True)
        
        if not flag and not is_found[0]:
            print(f"The value {value} is not present in the Binary Search Tree!")
            
        return self
    
    def get_children(self, value, is_found=None, flag=False):
        if not self:
            print("The Binary Search Tree is empty!")
            return 
        
        if is_found is None:
            is_found = [False]
        
        if self.data == value:
            is_found[0] = True
            children = []
            if self.left:
                children.append(self.left.data)
            if self.right:
                children.append(self.right.data)
            
            if not children:
                print(f"Match found, but children not available for {value}")
            return children
        
        if value < self.data: 
            if self.left:
                children = self.left.get_children(value, is_found, flag=True)
                if children or is_found[0]:
                    if not children:
                        return
                    return children
        elif value > self.data:
            if self.right:
                children = self.right.get_children(value, is_found, flag=True)
                if children or is_found[0]:
                    if not children:
                        return
                    return children
        
        if not flag and not is_found[0]:    
            print(f"The value {value} is not present in the Binary Search Tree!")

    def inorder_traversal(self, elements=None):
        if not self:
            print("The Binary Search Tree is empty!")
            return 
        
        if elements is None:
            elements = []
        
        if self.left:
            self.left.inorder_traversal(elements)
            
        elements.append(self.data)
        
        if self.right:
            self.right.inorder_traversal(elements)
            
        return elements
    
    def preorder_traversal(self, elements=None):
        if not self:
            print("The Binary Search Tree is empty!")
            return 
        
        if elements is None:
            elements = []
            
        elements.append(self.data)
        
        if self.left:
            self.left.preorder_traversal(elements)
        
        if self.right:
            self.right.preorder_traversal(elements)
            
        return elements
        
    def postorder_traversal(self, elements=None):
        if not self:
            print("The Binary Search Tree is empty!")
            return 
        
        if elements is None:
            elements = []
        
        if self.left:
            self.left.postorder_traversal(elements)
        
        if self.right:
            self.right.postorder_traversal(elements)
            
        elements.append(self.data)
            
        return elements
    
    def get_min_value(self):
        if not self:
            print("The Binary Search Tree is empty!")
            return
        
        if self.left:
            return self.left.get_min_value()
        return self.data
    
    def get_max_value(self):
        if not self:
            print("The Binary Search Tree is empty!")
            return
        
        if self.right:
            return self.right.get_max_value()
        return self.data
    
    def get_left_sum(self):
        if not self:
            print("The Binary Search Tree is empty!")
            return 0
        
        sum = 0
        if self.left:
            return self.left.calculate_sum()
            
        return 0
    
    def get_right_sum(self):
        if not self:
            print("The Binary Search Tree is empty!")
            return 0
        
        if self.right:
            return self.right.calculate_sum()
            
        return 0
    
    def calculate_sum(self):
        if not self:
            print("The Binary Search Tree is empty!")
            return 0
        
        left, right = 0, 0
        if self.left:
            left += self.left.calculate_sum()
            
        if self.right:
            right += self.right.calculate_sum()
            
        return left + self.data + right
    
    def get_depth(self):
        if not self:
            print("The Binary Search Tree is empty!")
            return 0
        
        ldepth, rdepth = 0, 0
        if self.left:
            ldepth += self.left.get_depth()
        
        if self.right:
            rdepth += self.right.get_depth()
            
        return max(ldepth, rdepth) + 1

    def get_height(self):
        if not self:
            print("The Binary Search Tree is empty!")
            return 0
        
        lheight, rheight = 0, 0   
        if self.left:
            lheight += self.left.get_depth()
        
        if self.right:
            rheight += self.right.get_depth()
            
        return max(lheight, rheight) + 1

    def depth_of(self, value, depth=None, flag=False):
        if not self:
            print("The Binary Search Tree is empty!")
            return 0
        
        if depth is None:
            depth = 0
        
        if self.data == value:
            return depth
        
        if value < self.data:
            if self.left:
                depth += 1
                depth = self.left.depth_of(value, depth, flag=True)
                if depth:
                    return depth
        elif value > self.data:
            if self.right:
                depth += 1
                depth = self.right.depth_of(value, depth, flag=True)
                if depth:
                    return depth

        if not flag:
            print(f"The value {value} is not present in the Binary Search Tree!")

    def height_of(self, value, flag=False):
        if not self:
            print("The Binary Search Tree is empty!")
            return 0
    
        if self.data == value:
            return self.get_height() - 1
        
        if value < self.data:
            if self.left:
                height = self.left.height_of(value, flag=True)
                if isinstance(height, int):
                    return height
        elif value > self.data:
            if self.right:
                height = self.right.height_of(value, flag=True)
                if isinstance(height, int):
                    return height
            
        if not flag:
            print(f"The value {value} is not present in the Binary Search Tree!")

    def get_diameter(self):
        if not self:
            print("The Binary Search Tree is empty!")
            return 0
        
        ldepth, rdepth = 0, 0 
        ldiameter, rdiameter = 0, 0
        if self.left:
            ldepth = self.left.get_depth()
            ldiameter = self.left.get_diameter()
            
        if self.right:
            rdepth = self.right.get_depth()
            rdiameter = self.right.get_diameter()
            
        return max(ldepth + rdepth + 1, max(ldiameter, rdiameter))
    
    def diameter_of(self, value, flag=False):
        if not self:
            print("The Binary Search Tree is empty!")
            return 0
        
        if self.data == value:
            return self.get_diameter()
        
        if value < self.data:
            if self.left:
                diameter = self.left.diameter_of(value, flag=True)
                if diameter:
                    return diameter
        elif value > self.data:
            if self.right:
                diameter = self.right.diameter_of(value, flag=True)
                if diameter:
                    return diameter
        
        if not flag:
            print(f"The value {value} is not present in the Binary Search Tree!")
    
    def dfs_traversal(self):
        return self.inorder_traversal()
        
    def bfs_traversal(self):
        if not self:
            print("The Binary Search Tree is empty!")
            return
        
        data = []
        items = [self]
        while items:
            for _ in items:
                node = items.pop(0)
                data.append(node.data)
                
                if node.left:
                    items.append(node.left)

                if node.right:
                    items.append(node.right)
                    
        return data
    
    def get_parent(self, value, flag=False):
        if not self:
            print("The Binary Search Tree is empty!")
            return
        
        if self.get_root(validate=True) == value:
            print(f"The value {value} is the root of the Binary Search Tree and does not have a parent!")
            return
        
        if self.left:
            if self.left.data == value:
                return self.data
        if self.right:
            if self.right.data == value:
                return self.data
        
        if value < self.data:
            if self.left:
                parent = self.left.get_parent(value, flag=True)
                if parent:
                    return parent
        elif value > self.data:
            if self.right:
                parent = self.right.get_parent(value, flag=True)  
                if parent:
                    return parent
        
        if not flag:
            print(f"The value {value} is not present in the Binary Search Tree!")
    
    def get_grand_children(self, value, is_children=None, is_found=None, flag=False):
        if not self:
            print("The Binary Search Tree is empty!")
            return
        
        if is_children is None and is_found is None:
            is_found = [False]
            is_children = [False]
        
        if self.data == value:
            is_found[0] = True
            grandchildren = []
            if self.left:
                is_children[0] = True
                if self.left.left:
                    grandchildren.append(self.left.left.data)
                if self.left.right:
                    grandchildren.append(self.left.right.data)
            if self.right:
                is_children[0] = True
                if self.right.left:
                    grandchildren.append(self.right.left.data)
                if self.right.right:
                    grandchildren.append(self.right.right.data)
            
            if not grandchildren:
                if is_children[0]:
                    print("Match found, but no grandchildren found!")
                    return grandchildren
                
                print("Match found, but no children and grandchildren found!")
                return grandchildren
                
            return grandchildren
                
        if value < self.data:
            if self.left:
                grandchildren = self.left.get_grand_children(value, is_children, is_found, flag=True) 
                if grandchildren or is_found[0]:
                    if not grandchildren:
                        return
                    return grandchildren
                    
        elif value > self.data:
            if self.right:
                grandchildren = self.right.get_grand_children(value, is_children, is_found, flag=True)
                if grandchildren or is_found[0]:
                    if not grandchildren:
                        return
                    return grandchildren
            
        if not flag and not is_found[0]:
            print(f"The value {value} is not present in the Binary Search Tree!")
                
    
    def get_grand_parent(self, value, flag=False):
        if not self:
            print("The Binary Search Tree is empty!")
            return
        
        if self.get_root(validate=True) == value:
            print(f"The value {value} is the root of the Binary Search Tree and does not have a parent and grand parent!")
            return
        
        if self.left:
            if self.left.get_root(validate=True):
                if self.left.data == value:
                    print(f"The parent of {value} is the root of the Binary Search Tree and does not have a grand parent!")
                    return
                
        if self.right:
            if self.right.get_root(validate=True):
                if self.right.data == value:
                    print(f"The parent of {value} is the root of the Binary Search Tree and does not have a grand parent!")
                    return
        
        if self.left:
            if self.left.left:
                if self.left.left.data == value:
                    return self.data
            if self.left.right:
                if self.left.right.data == value:
                    return self.data
        elif self.right:
            if self.right.right:
                if self.right.right.data == value:
                    return self.data
            if self.right.left:     
                if self.right.left.data == value:
                    return self.data   
                    
        if value < self.data:
            if self.left:
                val = self.left.get_grand_parent(value, flag=True)
                if val:
                    return val
        elif value > self.data:
            if self.right:
                val = self.right.get_grand_parent(value, flag=True)
                if val:
                    return val
        
        if not flag:
            print(f"The value {value} is not present in the Binary Search Tree!")

    def get_leaf_nodes(self, leaves=None):
        if not self:
            print("The Binary Search Tree is empty!")
            return
        
        if leaves is None:
            leaves = []
            
        if self.left:
            if self.left.left is None and self.left.right is None:
                leaves.append(self.left.data)
        if self.right:
            if self.right.left is None and self.right.right is None:
                leaves.append(self.right.data)
            
        if self.left:
            self.left.get_leaf_nodes(leaves)

        if self.right:
            self.right.get_leaf_nodes(leaves)
        
        return leaves
    
def build_tree(numbers):
    root = BST(numbers[0])
    root.parent = True
    for i in numbers[1:]:
        root.add_child(i)
    return root
```
