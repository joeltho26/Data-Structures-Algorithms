```python
from typing import Optional
class Node:
    def __init__(self, data, next=None):
        self.data = data
        self.next = next
        
class LinkedList:
    def __init__(self):
        self.head, self.rear = None, None
        self.count = 0
    
    def print_forward(self):
        itr = self.head
        ff = ""
        while itr.next:
            ff += str(itr.data) + ", "
            itr = itr.next
        ff += str(itr.data)
        return ff
        
    def insert_at_beginning(self, value):
        if self.head is None:
            self.rear = self.head = Node(value)
            self.count += 1
            return
    
        self.head = Node(value, self.head)
        self.count += 1
           
    def insert_at_end(self, value):
        if self.head is None:
            self.rear = self.head = Node(value)
            self.count += 1
            return   
        
        itr = self.head
        while itr.next:
            itr = itr.next
        itr.next = Node(value)
        self.rear = itr.next
        self.count += 1
        
    def search(self, value):
        itr = self.head
        while itr:
            if itr.data == value:
                return True
            itr = itr.next
        return False 
            
    def insert_at(self, index, value):
        is_invalid = False
        if index < 0 or index > self.count:
            print(f"Invalid index {index}!")
            is_invalid = True
            return is_invalid
        
        if index == 0:
            self.insert_at_beginning(value)
            return is_invalid
        
        count = 0
        itr = self.head
        while itr:
            if count == index - 1:
                itr.next = Node(value, itr.next)
                self.count += 1
                if self.count-1 == index:
                    self.rear = itr.next
                return is_invalid
            itr = itr.next
            count += 1 
            
    def delete_at_beginning(self):
        val = self.head.data
        self.head = self.head.next
        if self.head is None:
            self.rear = self.head
        self.count -= 1
        return val

    def delete_at_end(self):
        if self.count == 1:
            return self.delete_at_beginning()
        
        itr = self.head
        while itr.next.next:
            itr = itr.next
        val = itr.next.data
        itr.next = None
        self.rear = itr
        self.count -= 1
        return val

    def delete_by_value(self, value):
        if self.head.data == value:
            return self.delete_at_beginning()
            
        itr = self.head
        while itr.next:
            if itr.next.data == value:
                if itr.next == self.rear:
                    self.rear = itr
                itr.next = itr.next.next
                self.count -= 1
                return value
            itr = itr.next

    def delete_at(self, index):
        if index < 0 or index >= self.count:
            print(f"Invalid Index {index}!")
            return
        
        if index == 0:
            return self.delete_at_beginning()
        
        count = 0
        itr = self.head
        while itr:
            if count == index - 1:
                if self.count-2 == count:
                    self.rear = itr
                val = itr.next.data
                itr.next = itr.next.next
                self.count -= 1
                return val
            count += 1
            itr = itr.next
    
    def rotate_right(self, k):
        rotation = k%self.count
        for _ in range(rotation):
            val = self.delete_at_end()
            self.insert_at_beginning(val)

    def rotate_left(self, k):
        rotation = k%self.count
        for _ in range(rotation):
            val = self.delete_at_beginning()
            self.insert_at_end(val)

    def reverse(self):
        itr = self.head
        self.head = None
        while itr:
            self.insert_at_beginning(itr.data)
            itr = itr.next 

    def sort(self):
        swapped = True
        while swapped:
            swapped = False
            itr = self.head
            while itr.next:
                if itr.data > itr.next.data:
                    itr.data, itr.next.data = itr.next.data, itr.data
                    swapped = True
                itr = itr.next

            if not swapped:
                break
            
    def last_item(self):
        return self.rear
        
class Queue:
    def __init__(self, size):
        self.size = size
        self.container = LinkedList()
        
    def enqueue(self, value, index: Optional[int]=None, rotation: Optional[bool]=False):
        if not self.is_full():
            if index is None:
                self.container.insert_at_end(value)
                self.sort(flag=True)
                if not rotation:
                    print(f"{value} added to the queue!")
                return True
            else:
                flag = self.container.insert_at(index, value)
                self.sort(flag=True)
                if not flag:
                    if not rotation:
                        print(f"{value} added to the queue via index {index}!")
                    return True
                else:
                    if not rotation:
                        print(f"Unable to add {value} to the queue!")
                    
        return False

    def dequeue(self, index: Optional[int]=None, rotation: Optional[bool]=False):
        if not self.is_empty():
            if index is None:
                deleted_val = self.container.delete_at_beginning()
                if not rotation:
                    print(f"{deleted_val} removed from the queue!")
                return deleted_val
            else:
                deleted_val = self.container.delete_at(index)
                if deleted_val is not None:
                    if not rotation:
                        print(f"{deleted_val} removed from the queue via {index}!")
                    return deleted_val
                else:
                    if not rotation:
                        print(f"Unable to remove item from the queue via {index}!")
    
    def get_size(self):
        return self.size
    
    def get_count(self):
        if not self.is_empty():
            return self.container.count
        return 0

    def peek(self):
        if not self.is_empty():
            return "Peek: " + str(self.container.last_item().data)
        
    def first_item(self):
        if not self.is_empty():
            return self.container.head.data

    def remove(self, value):
        if not self.is_empty():
            if self.search(value, validate=True):
                self.container.delete_by_value(value)
                print(f"{value} removed from the queue!")
                return value
            else:
                print(f"{value} not found in the queue to remove!")

    def display(self):
        if not self.is_empty():
            print("Items in the queue are: {0}".format(self.container.print_forward()))

    def enqueue_values(self, values):
        if not self.is_full():
            flag = True
            for index, i in enumerate(values):
                flag = self.enqueue(i)
                if not flag:
                    print(f"Items unable to add into the queue are: {values[index:]}")

    def search(self, value, validate:Optional[bool]=False):
        if not self.is_empty():
            if self.container.search(value):
                if not validate:
                    print(f"{value} found in the queue!")
                return True
            else:
                if not validate:
                    print(f"{value} not found in the queue!") 
        return False

    def rotate_left(self, times):
        if not self.is_empty():
            self.container.rotate_left(times)
        print(f"The queue is rotated-left {times} times successfully!")

    def rotate_right(self, times):
        if not self.is_empty():
            self.container.rotate_right(times)
        print(f"The queue is rotated-right {times} times successfully!")
    
    def sort(self, flag: Optional[bool]=False):
        if not self.is_empty():
            self.container.sort()
            
        if not flag:
            print(f"The queue is sorted successfully!")

    def reverse(self):
        if not self.is_empty():
            self.container.reverse()
        print(f"The queue is reversed successfully!")

    def is_empty(self):
        if self.container.count == 0:
            print("The queue is empty!")
            return  True
        return False

    def is_full(self):
        if self.container.count == self.size:
            print("The queue is full!")
            return  True
        return False
```

```python
if __name__ == "__main__":
    # stack push/pop on left size 
    s = Queue(5)
    s.enqueue(5)
    s.enqueue(10)
    s.enqueue(15)
    s.enqueue(20, 1)
    s.enqueue(25, 3)
    s.display()
    
```
