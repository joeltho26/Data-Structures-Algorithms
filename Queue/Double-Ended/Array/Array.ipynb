```python
from typing import Optional
from collections import deque

class Queue:
    def __init__(self, size):
        self.size = size
        self.container = deque()
        
    def enqueue(self, value, index: Optional[int]=None, rotation: Optional[bool]=False):
        if not self.is_full():
            if index is None:
                self.container.append(value)
                if not rotation:
                    print(f"{value} is added in the queue!")
            else:
                if index > len(self.container):
                    print(f"Invalid index {index}!")
                    return False
                
                self.container.insert(index, value)
                if not rotation:
                    print(f"{value} is added in the queue via index {index}!")
            return True
            
        return False
    
    def enqueue_left(self, value, rotation: Optional[bool]=False):
        if not self.is_full():
            self.container.appendleft(value)
            if not rotation:
                print(f"{value} is added in the queue!")
            return True
            
        return False
    
    def dequeue(self, index: Optional[int]=None, rotation: Optional[bool]=False):
        if not self.is_empty():
            deleted_val = None
            if index is None:
                deleted_val = self.container.popleft()
                if not rotation:
                    print(f"{deleted_val} is deleted from the queue!")
            else:
                if index >= len(self.container):
                    print(f"Invalid index {index}!")
                    return
                deleted_val = self.container[index]
                self.container.remove(deleted_val)
                if not rotation:
                    print(f"{deleted_val} is deleted from the queue via index {index}!")
                
            if deleted_val:
                return deleted_val
            
    def dequeue_right(self, rotation: Optional[bool]=False):
        if not self.is_empty():
            deleted_val = self.container.pop()
            if not rotation:
                print(f"{deleted_val} is deleted from the queue!")
            return deleted_val
            
    def display(self):
        if not self.is_empty():
            print(f"The items in the queue are:", end=" ")
            for i in range(len(self.container)-1):
                print(self.container[i], end=", ")
            print(self.container[-1])
        
    def remove(self, value):
        if not self.is_empty():
            if value in self.container:
                self.container.remove(value)
                print(f"{value} is deleted from the queue!")
                return value
            else:
                print("{value} not found in queue to remove!")
    
    def search(self, value):
        if not self.is_empty():
            if value in self.container:
                print(f"{value} found in queue")
                return True
            
            print(f"{value} not found in queue")
        return False
        
    def get_count(self):
        if not self.is_empty():
            return len(self.container)
        return 0
    
    def get_size(self, rotation: Optional[bool]=False):
        if not rotation:
            print(f"The size of the queue is {self.size}")
        return self.size
    
    def peek(self):
        if not self.is_empty():
            return "Peek:" + str(self.container[-1])
    
    def enqueue_values(self, values):
        if not self.is_full():
            for i in range(len(values)):
                flag = self.enqueue(values[i])
                if not flag:
                    print(f"{values[i:]} were unable to be added into the queue!")
                    break
    
    def rotate_left(self, k):
        if not self.is_empty():
            rotation = k%self.size
            for _ in range(rotation):
                val = self.dequeue(index=0, rotation=True)
                self.enqueue(value=val, rotation=True)
            print(f"The queue has been rotated-left {k} times successfully!")
    
    def rotate_right(self, k):
        if not self.is_empty():
            rotation = k%self.size
            for _ in range(rotation):
                val = self.dequeue(rotation=True)
                self.enqueue(value=val, index=0, rotation=True)
            print(f"The queue has been rotated-right {k} times successfully!")
    
    def first_item(self):
        if not self.is_empty():
            return self.container[0]
    
    def sort(self):
        if not self.is_empty():
            size = len(self.container)
            for i in range(size):
                swapped = False
                for j in range(0, size-i-1):
                    if self.container[j] > self.container[j+1]:
                        self.container[j], self.container[j+1] = self.container[j+1], self.container[j]
                        swapped = True
                        
                if not swapped:
                    break
                
            print("Items in the queue are sorted in ascending order!")
    
    def reverse(self):
        if not self.is_empty():
            self.container.reverse()
            print(f"The queue has been reversed successfully!")
    
    def is_empty(self, validate: Optional[bool]=False):
        if len(self.container) == 0:
            if not validate:
                print("The queue is empty!")
            return True
        return False
    
    def is_full(self, validate: Optional[bool]=False):
        if len(self.container) == self.size:
            if not validate:
                print("The queue is full!")
            return True
        return False
    
```

```python
if __name__ == "__main__":
    s = Queue(5)
    s.enqueue(5)
    s.enqueue_left(10)
    s.display()
    s.enqueue(15)
    s.enqueue(20, 1)
    s.enqueue(1,10)
    s.enqueue_left(25)
    s.display()
```

```python
print("Count of items:", s.get_count())
print(s.is_full())
print(s.is_empty())
```

```python
print(s.dequeue())
print(s.dequeue_right())
print(s.dequeue(3))
print(s.peek())
s.display()
```

```python
print(s.remove(20))
s.display()
s.enqueue(7, 2)
s.display()
```

```python
s.enqueue_values((15, 4, 11, 21, 55))
s.display()
```

```python
print(s.search(5))
print(s.search(25))
```

```python
s.rotate_left(3)
s.display()
s.rotate_right(2)
s.display()
print("First item:", s.first_item())
```
