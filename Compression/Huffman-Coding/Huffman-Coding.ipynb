```python
class Node:
    def __init__(self, char=None, freq=0):
        self.char = char
        self.freq = freq
        self.left = None
        self.right = None
        
class HuffmanCoding:
    def calculate_frequencies(self, word):
        frequencies = {}
        for char in word:
            if char not in frequencies:
                freq = word.count(char)
                frequencies[char] = freq
                nodes.append(Node(char, freq))
                
    def build_huffman_tree(self):
        while len(nodes) > 1:
            nodes.sort(key=lambda x: x.freq)
            left = nodes.pop(0)
            right = nodes.pop(0)
            merged = Node(freq=left.freq + right.freq)
            merged.left = left
            merged.right = right
            nodes.append(merged)
        return nodes[0]
    
    def huffman_codes(self, node, current_code):
        if not node:
            return 
        
        if node.char:
            codes[node.char] = current_code
            
        self.huffman_codes(node.left, current_code+"0")
        self.huffman_codes(node.right, current_code+"1")

    def generate_huffman_codes(self, word):
        self.calculate_frequencies(word)
        root = self.build_huffman_tree()
        self.huffman_codes(root, "")
        return "".join(codes[char] for char in word)
        

    def decode_huffman_code(self, encoded_value):
        decoded_value = []
        current_code = ""
        code_to_char = {v:k for k,v in codes.items()}
        for bit in encoded_value:
            current_code += bit
            if current_code in code_to_char:
                decoded_value.append(code_to_char[current_code])
                current_code = ""
        return "".join(i for i in decoded_value)
        
```
