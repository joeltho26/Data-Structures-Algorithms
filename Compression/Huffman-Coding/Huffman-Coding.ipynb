```python
class Node:
    def __init__(self, char=None, freq=0):
        self.left = None
        self.right = None
        self.char = char
        self.freq = freq

class HuffmanCoding:
    def calculate_frequencies(self, nodes, word):
        frequencies = {}
        for char in word:
            if char not in frequencies:
                freq = word.count(char)
                frequencies[char] = freq
                nodes.append(Node(char=char, freq=freq))

    def build_huffman_tree(self, nodes):
        while len(nodes) > 1:
            nodes.sort(key=lambda x: x.freq)
            left = nodes.pop(0)
            right = nodes.pop(0)
            merged = Node(freq=left.freq + right.freq)
            merged.left = left
            merged.right = right
            nodes.append(merged)
        return nodes[0]
            
    def generate_huffman_coding(self, node, current_code, codes):
        if not node:
            return 
        
        if node.char:
            codes[node.char] = current_code
            
        self.generate_huffman_coding(node.left, current_code+"0", codes)
        self.generate_huffman_coding(node.right, current_code+"1", codes)
    
class HuffmanDecoding:
    def decode_huffman_code(self, code_to_char, encoded_value):
        current_code = ""
        decoded_value = []
        for bit in encoded_value:
            current_code += bit
            if current_code in code_to_char:
                decoded_value.append(code_to_char[current_code])
                current_code = ""
        return "".join(i for i in decoded_value)
    
def huffman_encoding(word):
    encoding = HuffmanCoding()
    encoding.calculate_frequencies(nodes, word)
    root = encoding.build_huffman_tree(nodes)
    encoding.generate_huffman_coding(root, "", codes)
    encoded_value = "".join(codes[char] for char in word)
    return encoded_value

def huffman_decoding(codes, encoded_value):
    code_to_char = {v:k for k,v in codes.items()}
    decoding = HuffmanDecoding()
    decoded_value = decoding.decode_huffman_code(code_to_char, encoded_value)
    return decoded_value
```
