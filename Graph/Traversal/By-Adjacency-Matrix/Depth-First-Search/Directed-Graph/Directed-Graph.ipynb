```python
class Graph:
    def __init__(self, size):
        self.size = size
        self.adj_matrices = [[0] * self.size for _ in range(self.size)]
        self.vertex_data = [''] * self.size
        self.visited = [False] * self.size

    def add_edges(self, u, v):
        if 0 <= u < self.size and 0 <= v < self.size:
            self.adj_matrices[u][v] = 1
        
    def add_vertex(self, index, vertex):
        if 0 <= index < self.size:
            self.vertex_data[index] = vertex
            
    def dfs(self, start_vertex, is_first=True):
        if is_first:
            print("Depth first search of the graph:")
        start_vertex_index = self.vertex_data.index(start_vertex)
        self.visited[start_vertex_index] = True
        print(f"{start_vertex}", end="-->")
        for i in range(self.size):
            if self.adj_matrices[start_vertex_index][i] == 1 and not self.visited[i]:
                self.dfs(self.vertex_data[i], is_first=False)
        
    def print_graph(self):
        print("The adjacent matrix of the graph:")
        print(" " * 2, end="")
        for i in range(self.size):
            print(self.vertex_data[i], end=" ")
        print()
        for i, row in enumerate(self.adj_matrices):
            print(self.vertex_data[i], end=" ")
            print(" ".join(map(str,row)))
            
    def print_vertex(self):
        print("The vertex of the graph:")
        print(self.vertex_data)
```

```python
if __name__ == "__main__":
    vertices = ['A','B','C','D','E','F','G']
    vertices_data = [('D','A'),('E','A'), ('C','E'),
                    ('A','C'),('F','B'),('C','F'),
                    ('B','C'),('C','G')]
    size = len(vertices)
    graph = Graph(size)
    for i in range(len(vertices)):
        graph.add_vertex(i, vertices[i])
        
    for i in vertices_data:
        if isinstance(i, tuple) and len(i) == 2:
            source = graph.vertex_data.index(i[0])
            destination = graph.vertex_data.index(i[1])
            graph.add_edges(source, destination)
            
    graph.print_graph()
    graph.print_vertex()
    graph.dfs('D')
    
```
