```python
from typing import Optional

class Hash:
    def __init__(self, size):
        self.size = size
        self.container = [None for _ in range(self.size)]
        # self.container = [None] * self.size
        
    def _generate_hash_key(self, data):
        key = 0
        for char in data:
            key += ord(char)
        return key%self.size
    
    def __getitem__(self, key: str):
        if not self.is_empty(validate=True):
            hash_key = self._generate_hash_key(key)
            if self.container[hash_key]:
                if isinstance(self.container[hash_key], tuple):
                    if len(self.container[hash_key]) == 2:
                        if self.container[hash_key][0] == key:
                            print(f"Item found: {self.container[hash_key]} at {hash_key}")
                            return self.container[hash_key]
            
            return f"No match found for {key}!"
    
    def __setitem__(self, key: str, value: int):
        if not self.is_full(validate=True):
            hash_key = self._generate_hash_key(key)
            if self.container[hash_key]:                
                if isinstance(self.container[hash_key], tuple):
                    if len(self.container[hash_key]) == 2:
                        if self.container[hash_key][0] == key:    
                            print(f"Updated item: {self.container[hash_key]} with {(key,value)} at {hash_key}!")
                            self.container[hash_key] = (key,value)
                            return
                        
                        print(f"Replaced item: {self.container[hash_key]} with {(key,value)} at {hash_key}")
                        self.container[hash_key] = (key,value)
                        return
                            
            print(f"Inserted item: {(key,value)} at {hash_key}")
            self.container[hash_key] = (key,value)
            return
    
    def __delitem__(self, key: str):
        if not self.is_empty(validate=True):
            hash_key = self._generate_hash_key(key)
            if isinstance(self.container[hash_key], tuple):
                if len(self.container[hash_key]) == 2:
                    if self.container[hash_key][0] == key:
                        val = self.container[hash_key]
                        print(f"Deleted item: {val} at {hash_key}!")
                        self.container[hash_key] = None
                        return val
            
            print("No match found!")
            
    def is_empty(self, validate: Optional[bool]=False):
        count = self.get_count()
        if count == 0:
            if validate:
                print("HashMap is empty!")
                return
            return True
        
        return False
    
    def is_full(self, validate: Optional[bool]=False):
        count = self.get_count()
        if count == self.size:
            if validate:
                print("HashMap is full!")
                return
            return True
        
        return False

    def get_count(self):
        count = 0 
        for data in self.container:
            if isinstance(data, list):
                for _ in data:
                    count += 1
            elif isinstance(data, tuple):
                count += 1
            
        return count
```
