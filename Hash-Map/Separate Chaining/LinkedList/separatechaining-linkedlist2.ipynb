```python
# Hashing - separate chaining with linked list

class Node:
    def __init__(self, data, next=None):
        self.data = data
        self.next = next
        
class LinkedList:
    def __init__(self):
        self.rear = self.head = None
        self.count = 0

    def insert_at_end(self, data):
        if self.head is None:
            self.rear = self.head = Node(data)
            self.count += 1
            print(f"- The Node {data} is inserted at the end successfully!")
            return

        self.rear.next = Node(data, self.rear.next)
        self.rear = self.rear.next
        self.count += 1
        print(f"- The Node {data} is inserted at the end successfully!")
        return
    
    def get_length(self):
        return self.count
            
    def search(self, key):
        itr = self.head
        count = 0
        while itr:
            if itr.data[0] == key:
                return True
            itr = itr.next
            count += 1
        return False
    
    def update_value(self, key, value):
        itr = self.head
        while itr:
            if itr.data[0] == key:
                itr.data = (key, value)
                print(f"- Updated key {key} with value {value} successfully!")
                return
            itr = itr.next
        print(f"- Key {key} not found to update!")
        return
    
    def get_value(self, key):
        if self.head is None:
            print("- Linked list is empty!")
            return

        itr = self.head
        while itr:
            if itr.data[0] == key:
                return itr.data[1]
            itr = itr.next
                
    def print_forward(self):
        if self.head is None:
            print("- Linked list is empty!")
            return
        
        itr = self.head
        fstr = ""
        while itr.next:
            fstr += str(itr.data) + "-->"
            itr = itr.next
        fstr += str(itr.data)
        print(fstr)
        
    def delete_by_value(self, key):
        if self.head is None:
            print("- Linked list is empty!")
            return
        
        if self.head.data[0] == key:
            val = self.head.data
            self.head = self.head.next
            self.count -= 1
            print(f"- The Node with key {key} is deleted successfully!")
            return val
        
        itr = self.head
        while itr.next:
            if itr.next.data[0] == key:
                print(f"- The Node with key {key} is deleted successfully!")
                val = itr.next.next
                self.count -= 1
                if itr.next == self.rear:
                    self.rear = itr
                    self.rear.next = None
                    return val
                itr.next = itr.next.next
                return val
            itr = itr.next
        
        print(f"- The Node with key {key} not found to delete!")
        return
    

class Hash:
    def __init__(self, size):
        self.size = size
        self.container = [None] * self.size
        
    def get_hash(self, key):
        h = 0
        for i in key:
            h += ord(i)
        return h%self.size
    
    def __setitem__(self, key, value):
        if not self.is_full():
            h = self.get_hash(key)
            if self.container[h]:
                if isinstance(self.container[h], LinkedList):
                    if self.container[h].search(key):
                        self.container[h].update_value(key, value)
                        return
                if isinstance(self.container[h], tuple):
                    if self.container[h][0] == key:
                        self.container[h] = (key, value)
                        print(f"Updated key {key} with value {value} successfully!")
                        return
                    else:
                        ll = LinkedList()
                        ll.insert_at_end(self.container[h])
                        ll.insert_at_end((key, value))
                        self.container[h] = ll
                        print("Created Linked list with values!")
                        return
                        
            self.container[h] = (key, value)
            return

        print("Hash Map is full!")
        return

    def __getitem__(self, key):
        if not self.is_empty():
            h = self.get_hash(key)
            if self.container[h]:
                if isinstance(self.container[h], LinkedList):
                    if self.container[h].search(key):
                        val = self.container[h].get_value(key)
                        print(f"The value for key {key} is {val}")
                        return val
                if isinstance(self.container[h], tuple):
                    if self.container[h][0] == key:
                        print(f"The value for key {key} is {self.container[h][1]}")
                        return self.container[h][1]
                    
            return f"Key {key} not found to get!"
        
        print("Hash Map is empty!")
        return

    def __delitem__(self, key):
        if not self.is_empty():
            h = self.get_hash(key)
            if self.container[h]:
                if isinstance(self.container[h], LinkedList):
                    if self.container[h].search(key):
                        val = self.container[h].delete_by_value(key)
                        print(f"Key {key} deleted successfully from linked list!")
                        print(self.container[h].get_length())
                        if self.container[h].get_length() == 1:
                            self.container[h] = (self.container[h].head.data[0], self.container[h].head.data[1])
                            print("Deleting Linked list with values!")
                        return val
                    
                if isinstance(self.container[h], tuple):
                    if self.container[h][0] == key:
                        val = self.container[h]
                        self.container[h] = None
                        print(f"Key {key} deleted successfully!")
                        return val
                        
                
            print(f"Key {key} not found to delete!")
            return
                
        print("Hash Map is empty!")
        return

    def is_empty(self):
        if self.get_count() == 0:
            return True
        return False

    def is_full(self):
        for i in self.container:
            if i is None:
                return False
        return True

    def get_count(self):
        count = 0
        for i in range(len(self.container)):
            if self.container[i] is not None:
                if isinstance(self.container[i], tuple):
                    count += 1
                if isinstance(self.container[i], LinkedList):
                    count += self.container[i].get_length()
        return count
    
    def print_linkedlist(self):
        ll = False
        for idx, i in enumerate(self.container):
            if isinstance(i, LinkedList):
                print(f"Index {idx} has linked list:")
                i.print_forward()
                ll = True
        
        if not ll:
            print("No linked list in the hash map!")
```
