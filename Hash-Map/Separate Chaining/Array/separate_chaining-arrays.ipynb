```python
# Hashing - Array included when there more than 1 element - Separate Chaining

class Hash:
    def __init__(self, size):
        self.size = size
        self.container = [None] * self.size
        
    def get_hash(self, key):
        h = 0
        for i in key:
            h += ord(i)
        return h%self.size
    
    def __setitem__(self, key, value):
        if not self.is_full():
            h = self.get_hash(key)
            if self.container[h]:
                if not isinstance(self.container[h], list):
                    if len(self.container[h]) == 2 and isinstance(self.container[h], tuple):
                        if self.container[h][0] == key:
                            self.container[h] = (key, value)
                            print(f"Updated value {value} for key {key}")
                            return
                        self.container[h] = [self.container[h]]
                        self.container[h].append((key,value))
                        print("Creating the list...")
                        print(f"Inserted Item: ({key,value}) into the list!")
                        return
                self.container[h].append((key, value))
                print(f"Inserted Item: ({key,value}) into the list!")
                return
            self.container[h] = (key, value)
            print(f"Inserted Item: ({key,value}) at {h}!")
            return

        print("Hash Map is full!")
        return

    def __getitem__(self, key):
        if not self.is_empty():
            h = self.get_hash(key)
            if self.container[h]:
                if isinstance(self.container[h], tuple) and len(self.container[h]) == 2:
                    if self.container[h][0] == key:
                        print(f"Item found: ({key, self.container[h][1]} at {h})")
                        return self.container[h][1]
                
                if isinstance(self.container[h], list):
                    for i in range(len(self.container[h])):
                        if self.container[h][i][0] == key:
                            print(f"Item found: ({key, self.container[h][i][1]} at {h} from list at {i})")
                            return self.container[h][i]
                    
            return f"Key {key} not found to get!"
        
        print("Hash Map is empty!")
        return

    def __delitem__(self, key):
        if not self.is_empty():
            h = self.get_hash(key)
            if self.container[h]:
                if isinstance(self.container[h], tuple) and len(self.container[h]) == 2:
                    if self.container[h][0] == key:
                        val = self.container[h]
                        self.container[h] = None
                        print(f"Deleted item: {val}")
                        return
                
                if isinstance(self.container[h], list):
                    for i in range(len(self.container[h])):
                        if self.container[h][i][0] == key:
                            val = self.container[h][i]
                            self.container[h].pop(i)
                            print(f"Deleted item: {val}")
                            if len(self.container[h]) == 1:
                                self.container[h] = self.container[h][0]
                                print("Converting from list to tuple...")
                            return
                    
            print(f"Key {key} not found to delete!")
            return
                
        print("Hash Map is empty!")
        return

    def is_empty(self):
        if self.get_count() == 0:
            return True
        return False

    def is_full(self):
        for i in self.container:
            if i is None:
                return False
        return True

    def get_count(self):
        count = 0
        for i in self.container:
            if i is not None:
                if isinstance(i, tuple):
                    count += 1
                if isinstance(i, list):
                    for _ in i:
                        count += 1
        return count
```
