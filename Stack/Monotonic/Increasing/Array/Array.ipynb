```python
# monotonically increasing stack - lower values at front and higher values at end
from typing import Optional

class Stack:
    def __init__(self, size):
        self.size = size
        self.container = []    
        
    def push(self, value, index: Optional[int]=None, rotation: Optional[bool]=False):
        if not self.is_full():
            if index is None:
                self.container.append(value)
                if not rotation:
                    print(f"{value} is added in the stack!")
            else:
                if index > len(self.container):
                    print(f"Invalid index {index}!")
                    return False
                self.container.insert(index, value)
                if not rotation:
                    print(f"{value} is added in the stack via index {index}!")
            self.sort(property=True)
            return True
            
        return False
    
    def pop(self, index: Optional[int]=None, rotation: Optional[bool]=False):
        if not self.is_empty():
            deleted_val = None
            if index is None:
                deleted_val = self.container.pop(0)
                if not rotation:
                    print(f"{deleted_val} is deleted from the stack!")
            else:
                if index >= len(self.container):
                    print(f"Invalid index {index}!")
                    return
                deleted_val = self.container.pop(index)
                if not rotation:
                    print(f"{deleted_val} is deleted from the stack via index {index}!")
                
            if deleted_val:
                return deleted_val
    
    def display(self):
        if not self.is_empty():
            print(f"The items in the stack are:", end=" ")
            for i in range(len(self.container)-1):
                print(self.container[i], end=", ")
            print(self.container[-1])
        
    def remove(self, value):
        if not self.is_empty():
            if value in self.container:
                self.container.remove(value)
                print(f"{value} is deleted from the stack!")
                return value
            else:
                print("{value} not found in stack to remove!")
    
    def search(self, value):
        if not self.is_empty():
            if value in self.container:
                print(f"{value} found in stack")
                return True
            
            print(f"{value} not found in stack")
        return False
        
    
    def get_count(self):
        if not self.is_empty():
            return len(self.container)
        return 0
    
    def get_size(self, rotation: Optional[bool]=False):
        if not rotation:
            print(f"The size of the stack is {self.size}")
        return self.size
    
    def peek(self):
        if not self.is_empty():
            return "Peek:" + str(self.container[-1])
    
    def push_values(self, values):
        if not self.is_full():
            for i in range(len(values)):
                flag = self.push(values[i])
                if not flag:
                    print(f"{values[i:]} were unable to be added into the stack!")
                    break
    
    def rotate_left(self, k):
        if not self.is_empty():
            rotation = k%self.size
            self.container = self.container[rotation:] + self.container[:rotation]
            print(f"The stack has been rotated-left {k} times successfully!")
    
    def rotate_right(self, k):
        if not self.is_empty():
            rotation = k%self.size
            self.container = self.container[rotation+1:] + self.container[:rotation+1]
            print(f"The stack has been rotated-right {k} times successfully!")
    
    def first_item(self):
        if not self.is_empty():
            return self.container[0]
    
    def sort(self, property: Optional[bool]=False):
        if not self.is_empty():
            size = len(self.container)
            for i in range(size):
                swapped = False
                for j in range(0, size-i-1):
                    if self.container[j] > self.container[j+1]:
                        self.container[j], self.container[j+1] = self.container[j+1], self.container[j]
                        swapped = True
                        
                if not swapped:
                    break
                
            if not property:
                print("Items in the stack are sorted in ascending order!")
    
    def reverse(self):
        if not self.is_empty():
            self.container = self.container[::-1]
            if not property:
                print(f"The stack has been reversed successfully!")
    
    def is_empty(self, validate: Optional[bool]=False):
        if len(self.container) == 0:
            if not validate:
                print("The stack is empty!")
            return True
        return False
    
    def is_full(self, validate: Optional[bool]=False):
        if len(self.container) == self.size:
            if not validate:
                print("The stack is full!")
            return True
        return False
    
```

```python
if __name__ == "__main__":
    s = Stack(5)
    s.push(10)
    s.push(5)
    s.push(15)
    s.push(20, 1)
    s.push(1,10)
    s.push(25, 3)
    s.display()
```

```python
print("Count of items:", s.get_count())
print(s.is_full())
print(s.is_empty())
```

```python
print(s.pop())
print(s.pop(5))
print(s.pop(3))
print(s.peek())
s.display()
```

```python
print(s.remove(20))
s.display()
s.push(7, 2)
s.display()
```

```python
s.push_values((15, 4, 21))
s.display()
```

```python
print(s.search(5))
print(s.search(15))
```

```python
s.rotate_left(3)
s.display()
s.rotate_right(2)
s.display()
print("First item:", s.first_item())
```

```python
s.sort()
s.display()
s.reverse()
s.display()
s.get_size()
```
