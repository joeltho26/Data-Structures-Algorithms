```python
def armstrong(value):
    val = 0
    for i in str(value):
        val += int(i) ** 3
      
    if value == val:
        print(f"{value} is an armstrong number")
    else:
        print(f"{value} is not an armstrong number")

def main(values):
    for val in values:
        armstrong(val)
```

```python
values = [153, 274, 370]
main(values)
```
