```python
def armstrong(value):
    val = 0
    data = value
    while value:
        rem = value % 10
        value //= 10
        val += rem ** 3

    if data == val:
        print(f"{data} is an armstrong number")
    else:
        print(f"{data} is not an armstrong number")

def main(values):
    for val in values:
        armstrong(val)
```

```python
values = [153, 274, 370]
main(values)
```
