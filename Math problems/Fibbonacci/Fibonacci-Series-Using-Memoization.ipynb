```python
def fibonacci(value, memo):
    if memo[value]:
        return memo[value]
    if value <= 1:
        memo[value] = value
    else:
        memo[value] = fibonacci(value-1, memo) + fibonacci(value-2, memo)
    
    return memo[value]

def main(values):
    for val in values:
        memo = [None] * (val + 1)
        fibonacci(val, memo)
        print(f"Fibonacci({val}) memo table:", memo)
```

```python
if __name__ == "__main__":
    values = [4, 6, 15, 23]
    main(values)
```
