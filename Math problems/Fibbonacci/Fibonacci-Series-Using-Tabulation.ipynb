```python
def fibonacci(value, tab):
    if tab[value]:
        return tab[value]
    
    if value <= 1:
        tab[value] = value
        
    for i in range(2,value+1):
        tab[value] = fibonacci(i-1, tab) + fibonacci(i-2, tab)
    
    return tab[value]
    
def main(values):
    for val in values:
        tab = [None] * (val+1)
        fibonacci(val, tab)
        print(tab)
```

```python
if __name__ == "__main__":
    values = [4, 6, 15, 23]
    main(values)
```
