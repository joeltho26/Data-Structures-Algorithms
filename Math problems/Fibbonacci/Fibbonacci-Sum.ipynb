```python
def fibonacci(value):
    if value <= 1:
        return value
    else:
        return fibonacci(value-1) + fibonacci(value-2)
   
def main(values):
    for val in values:
        sum = fibonacci(val)
        print(sum)
```

```python
if __name__ == "__main__":
    values = [6, 15, 23]
    main(values)
```
