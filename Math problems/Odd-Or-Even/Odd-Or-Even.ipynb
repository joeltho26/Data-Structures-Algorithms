```python
def odd_or_even(value):
    if value > 0:
        if value%2 == 0:
            print(f"{value} is an even number!")
        elif value%2 == 1:
            print(f"{value} is an odd number!")
    else:
        print(f"{value} is neither even nor odd!")
        
def main(values):
    for val in values:
        odd_or_even(val)
```

```python
values = [12, 0, 9, 1]
main(values)
```
