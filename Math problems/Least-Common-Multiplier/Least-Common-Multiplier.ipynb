```python
def gcd(a, b):
    while b != 0:
        rem = a%b
        a = b
        b = rem
    return a

def lcm(a, b):
    return (a//gcd(a,b)*b)

def main(values):
    for val in values:
        count = len(val)
        k = lcm(val[0], val[1])
        if count > 2:
            times = count - 2
            for i in range(times):
                k = lcm(k, val[i+2])
        print(f"LCM of {val} is {k}")
```

```python
values = [(15, 12), (15, 12, 17), (240, 16, 196, 120)]
main(values)
```
