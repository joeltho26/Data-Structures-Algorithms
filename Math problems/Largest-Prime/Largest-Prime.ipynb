```python
def largest_prime(value):
    if value > 1:
        for i in range(value, 1, -1):
            flag = False
            for j in range(2, i):
                if i%j == 0:
                    flag=True
                    break

            if not flag:
                print(f"The largest prime is {i}")
                break
                

def main(values):
    for val in values:
        largest_prime(val)
```

```python
values = [100, 735, 1320]
main(values)
```
