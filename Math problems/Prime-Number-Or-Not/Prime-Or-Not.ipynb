```python
def prime(value):
    flag= False
    if value == 1:
        flag = True
    else:
        for i in range(2,value):
            if value%i == 0:
                flag = True
                break

    if flag:
        print(f"{value} is not a prime number")
    else:
        print(f"{value} is a prime number")
        

def main(numbers):
    for val in numbers:
        prime(val)
```

```python
numbers = [121, 7, 11, 12, 1]
main(numbers)
```
