```python
# approach 1
def palindrome(value):
    
    if value.lower() == value[::-1].lower():
        print(f"{value} is a palindrome")
    else:
        print(f"{value} is not a palindrome")
        
def main(values):
    for val in values:
        palindrome(val)
```

```python
if __name__ == "__main__":
    values = ["Malayalam", "Tamil", "Noon", "Loop", "a", "aa", "aab"]
    main(values)
```
