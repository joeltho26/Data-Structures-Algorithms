```python
# approach 2
def palindrome(value):
    size = len(value)
    limit = mid = int(size/2)
    if size%2 == 1:
        limit = mid + 1
    
    first = value[:mid].lower()
    last = "".join(list(reversed(value[limit:]))).lower()
    
    if first == last:
        print(f"{value} is a palindrome")
    else:
        print(f"{value} is not a palindrome")
        
def main(values):
    for val in values:
        palindrome(val)
```

```python
if __name__ == "__main__":
    values = ["Malayalam", "Tamil", "Noon", "Loop", "a", "aa", "aab"]
    main(values)
```
