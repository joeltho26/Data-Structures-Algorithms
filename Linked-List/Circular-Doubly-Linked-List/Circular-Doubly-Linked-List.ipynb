```python
# Circular Doubly Linked List
from typing import Optional

class Node:
    def __init__(self, data, prev=None, next=None):
        self.data = data
        self.next = next
        self.prev = prev
        
class LinkedList:
    def __init__(self):
        self.head = None
        
    def insert_at_beginning(self, data, rotation: Optional[bool]=False):
        if self.head is None:
            self.head = Node(data)
            self.head.next = self.head.prev = self.head
        else:
            fnode = self.head
            itr = fnode.prev
            node = Node(data,next=self.head, prev=itr)
            self.head.prev = node
            self.head = node
            itr.next = self.head
            
        if not rotation:
            print(f"Inserted node {data} at the beginning!")
    
    def insert_at_end(self, data, rotation: Optional[bool]=False):
        if self.head is None:
            self.head = Node(data)
            self.head.prev = self.head.next = self.head
        else:
            fnode = self.head
            itr = fnode.prev
            node = Node(data, prev=itr, next=fnode)
            itr.next = node
            self.head.prev = node
            
        if not rotation:
            print(f"Inserted node {data} at the end!")
        return

    def print_forward(self):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        itr = fnode = self.head
        fstr = ""
        while itr.next:
            if itr.next != fnode:
                fstr += str(itr.data) + "-->"
            elif itr.next == fnode:
                fstr += str(itr.data)
                break
            itr = itr.next
        if itr.next is None:
            fstr += str(itr.data)
        print(f"Printing elements forward: \n {fstr}")
        print("#"*25)

    def print_backward(self):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        itr = fnode = self.head
        bstr = ""
        while itr.next:
            if itr.next != fnode:
                bstr = "-->" + str(itr.data) + bstr 
            elif itr.next == fnode:
                bstr = str(itr.data) + bstr
                break        
            itr = itr.next
        if itr.next is None:
            bstr = str(itr.data) + bstr
        print(f"Printing elements backward: \n {bstr}")
        print("#"*25)

    def get_length(self):
        count = 0
        if self.head is None:
            return count
        
        itr = fnode = self.head
        while itr.next:
            if itr.next != fnode:
                count += 1
            else:
                break
            itr = itr.next
            
        if itr:
            count += 1
    
        return count

    def insert_values(self, data_list):
        for data in data_list:
            self.insert_at_end(data)
    
    def insert_at(self, index, data):
        if index < 0 or index > self.get_length():
            raise Exception("Invalid Index")
        
        if index == 0:
            self.insert_at_beginning(data, True)
        else:
            itr = fnode = self.head
            count = 0
            while itr.next != fnode:
                if count == index - 1:
                    node = Node(data, itr, itr.next)
                    if node.next:
                        node.next.prev = node
                    itr.next = node
                    break
                itr = itr.next
                count += 1
        
            if itr.next == fnode:
                if count == index - 1:
                    node = Node(data, itr, itr.next)
                    itr.next = node
                    self.head.prev = node
            
        print(f"Node {data} is inserted successfully at index {index}")

    def insert_after(self, value, data):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        is_inserted = False
        itr = fnode = self.head
        while itr.next != fnode:
            if itr.data == value:
                node = Node(data, itr, itr.next)
                if node.next:
                    node.next.prev = node
                itr.next = node
                is_inserted = True
            itr = itr.next
            
        if not is_inserted:
            if itr.next == fnode:  
                if itr.data == value:
                    node = Node(data, itr, itr.next)
                    itr.next = node
                    self.head.prev = node
                    is_inserted = True
            
        if is_inserted:
            print(f"Node {data} inserted after value {value} successfully!")
        else:
            print(f"No value {value} found to insert node after!")
    
    def insert_before(self, value, data):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        is_inserted = False
        if self.head.data == value:
            self.insert_at_beginning(data, True)
            is_inserted = True
        elif self.get_length() > 1:
            itr = fnode = self.head
            while itr.next != fnode:
                if itr.next.data == value:
                    node = Node(data, itr, itr.next)
                    if node.next:
                        node.next.prev = node
                    itr.next = node
                    is_inserted = True
                    break
                itr = itr.next
                
        if is_inserted:
            print(f"Node {data} inserted before value {value} successfully!")
        else:
            print(f"No value {value} found to insert node before!")

    def first_node(self):
        if self.head is None:
            print("Linked list is empty!")
            return
        print(f"Head node value: {self.head.data}")
        return self.head

    def last_node(self):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        itr = fnode = self.head
        while itr.next != fnode:
            itr = itr.next
        print(f"Last node value: {itr.data}")
        return itr

    def search(self, value):
        if self.head is None:
            print("Linked list is empty!")
            return False
        
        is_found = False
        itr = fnode = self.head
        while itr.next != fnode:
            if itr.data == value:
                is_found = True
                break
            itr = itr.next
           
        if itr.next == fnode: 
            if itr.data == value:
                is_found = True
            
        if is_found:
            print(f"Value {value} found!")
        else:
            print(f"Value {value} not found!")
        return is_found

    def delete_at_beginning(self, rotation: Optional[bool]=False):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        fnode = self.head
        value = self.head.data
        self.head = self.head.next
        if self.head == fnode:
            self.head = None
        else:
            fnode.prev.next = self.head
            self.head.prev = fnode.prev
        
        if not rotation:
            print(f"Node {value} at the beginning deleted!")
        return value

    def delete_at_end(self, rotation: Optional[bool]=False):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        value = None
        if self.get_length() == 1:
            value = self.head.data
            self.head = None

        if self.get_length() > 1:
            fnode = self.head
            itr = fnode.prev 
            value = itr.data
            itr.prev.next = fnode
            fnode.prev = itr.prev
            
        if not rotation:
            print(f"Node {value} at the end deleted!")
        return value
        
    def delete_at(self, index):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        if index < 0 or index >= self.get_length():
            raise Exception("Invalid Index")
        
        data = None
        if index == 0:
            data = self.delete_at_beginning(True)
        else:
            count = 0
            itr = fnode = self.head
            while itr.next != fnode:
                if count == index - 1:
                    data = itr.next.data
                    itr.next = itr.next.next
                    if itr.next != fnode:
                        itr.next.prev = itr
                    if itr.next == fnode:
                        self.head.prev = itr
                    break
                itr = itr.next
                count += 1
                
        print(f"Node {data} is deleted successfully at index {index}")
        return data

    def delete_by_value(self, value):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        is_deleted = False
        if self.head.data == value:
            self.delete_at_beginning(True)
            is_deleted = True
        else:
            itr = fnode = self.head
            while itr.next != fnode:
                if itr.next.data == value:
                    itr.next = itr.next.next
                    if itr.next != fnode:
                        itr.next.prev = itr
                    if itr.next == fnode:
                        self.head.prev = itr
                    is_deleted = True
                    break
                itr = itr.next
                
        if is_deleted:
            print(f"Node with value {value} deleted successfully!")
            return value
        else:  
            print(f"No value {value} found to delete node by value!")

    def delete_after(self, value):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        is_deleted = False
        data = None
        itr = fnode = self.head
        while itr.next != fnode:
            if itr.data == value:
                if itr.next:
                    data = itr.next.data
                    itr.next = itr.next.next
                    if itr.next:
                        itr.next.prev = itr
                is_deleted = True
                break
            itr = itr.next
        
        if not is_deleted:
            if itr.next == fnode:
                if itr.data == value:
                    data = self.delete_at_beginning(rotation=True)
                    is_deleted = True
                
        if is_deleted:
            print(f"Node {data} deleted after value {value} successfully!")
        else:
            print(f"No value {value} found to delete node after!")
        return data
    
    def delete_before(self, value):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        data = None
        is_deleted = False
        if self.head.data == value:
            data = self.delete_at_end(rotation=True)
            is_deleted = True
        elif self.head.next.data == value:
            data = self.delete_at_beginning(True)
            is_deleted = True
        else:
            itr = fnode = self.head
            while itr.next.next != fnode:
                if itr.next.next.data == value:
                    data = itr.next.data
                    itr.next = itr.next.next
                    if itr.next:
                        itr.next.prev = itr
                    is_deleted = True
                    break
                itr = itr.next
                
        if is_deleted:
            print(f"Node {data} deleted before value {value} successfully!")
        else:
            print(f"No value {value} found! to delete node!")
        
        return data
    
    def rotate_left(self, k):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        times = k%self.get_length()
        for _ in range(times):
            val = self.delete_at_beginning(True)
            self.insert_at_end(val, True)
        
        print(f"The list rotated {k} times to the left successfully!")
        return

    def rotate_right(self, k):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        times = k%self.get_length()
        for _ in range(times):
            val = self.delete_at_end(True)
            self.insert_at_beginning(val, True)
            
        print(f"The list rotated {k} times to the right successfully!")
        return

    def sort(self):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        swapped = True
        while swapped:
            swapped = False
            itr = fnode = self.head
            while itr.next != fnode:
                if itr.data > itr.next.data:
                    itr.data, itr.next.data = itr.next.data, itr.data
                    swapped = True
                itr = itr.next
                
            if not swapped:
                break
            
        print("The list is sorted successfully!")

    def reverse(self):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        itr = fnode = self.head
        self.head = None
        while itr.next != fnode:
            self.insert_at_beginning(itr.data, True)
            itr = itr.next
            
        if itr.next == fnode:
            self.insert_at_beginning(itr.data, True)
            
        print("The list is reversed successfully!")
    
    def detect_loop(self):    
        if self.head is None:
            print("Linked list is empty!")
            return False
        
        container = set()
        itr = self.head
        while itr:
            if itr in container:
                print("Loop exists!")
                #self.remove_loop(itr)
                return True
            container.add(itr)
            itr = itr.next
            
        print("No loop exist!")
        return False

    def floyd_cycle(self):
        if self.head is None:
            print("Linked list is empty!")
            return False
        
        slow_p = fast_p = self.head
        while fast_p and fast_p.next:
            slow_p = slow_p.next
            fast_p = fast_p.next.next
            if slow_p == fast_p:
                self.remove_loop(slow_p)
                print("Loop exists!")
                return True
            
        print("No loop exist!")
        return False

    def remove_loop(self, itr):
        # Approach 1
        # ptr1 = ptr2 = itr
        # k = 1
        # while ptr1 != ptr2.next:
        #     ptr2 = ptr2.next
        #     k += 1
            
        # ptr1 = ptr2 = self.head
        # for _ in range(k):
        #     ptr2 = ptr2.next
            
        # while ptr1 != ptr2:
        #     ptr1 = ptr1.next
        #     ptr2 = ptr2.next
            
        # while ptr1 != ptr2.next:
        #     ptr2 = ptr2.next
        # ptr2.next = None
        
        # approach 2
        # p2 = itr
        # p1 = self.head
            
        # while p1 != p2:
        #     p2 = p2.next
        #     p1 = p1.next
            
        # while p1 != p2.next:
        #     p2 = p2.next
        # p2.next = None
        
        # Approach 3
        p1 = p2 = itr
        while p1 != p2.next:
            p2 = p2.next
        p2.next = None
        
    def detect_loop_remove_loop_alternate(self):
        slow_p = fast_p = self.head
        while slow_p and fast_p and fast_p.next:
            slow_p = slow_p.next
            fast_p = fast_p.next.next
            if slow_p == fast_p:
                print("Loop exist!")
                slow_p = self.head
                while slow_p != fast_p:
                    slow_p = slow_p.next
                    fast_p = fast_p.next
            
                while slow_p != fast_p.next:
                    fast_p = fast_p.next
                fast_p.next = None
                print("Loops removed successfully!")
                return
        print("No loop exist!")
```
