```python
# Non circular Singly Linked List
from typing import Optional

class Node:
    def __init__(self, data, next=None):
        self.data = data
        self.next = next
        
class LinkedList:
    def __init__(self):
        self.head = None
        
    def insert_at_beginning(self, data, rotation: Optional[bool]=False):
        if self.head is None:
            self.head = Node(data, None)
        else:
            self.head = Node(data, self.head)
            
        if not rotation:
            print(f"Node {data} inserted at the beginning successfully!")
    
    def insert_at_end(self, data, rotation: Optional[bool]=False):
        if self.head is None:
            self.head = Node(data, None)
        else:
            itr = self.head
            while itr.next:
                itr = itr.next
            itr.next = Node(data, itr.next)
            
        if not rotation:
            print(f"Node {data} inserted at the end successfully!")

    def print_forward(self):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        itr = self.head
        fstr = ""
        while itr.next:
            fstr += str(itr.data) + "-->"
            itr = itr.next
        fstr += str(itr.data)
        print(f"Printing elements forward: \n {fstr}")
        print("#"*25)

    def print_backward(self):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        itr = self.head
        bstr = ""
        while itr.next:
            bstr = "-->" + str(itr.data) + bstr 
            itr = itr.next
        bstr = str(itr.data) + bstr
        print(f"Printing elements backward: \n {bstr}")
        print("#"*25)

    def get_length(self):
        count = 0
        itr = self.head
        while itr:
            count += 1
            itr = itr.next
        return count

    def insert_values(self, data_list):
        for data in data_list:
            self.insert_at_end(data)
    
    def insert_at(self, index, data):
        if index < 0 or index > self.get_length():
            raise Exception("Invalid Index")
        
        if index == 0:
            self.insert_at_beginning(data, True)
        else:
            itr = self.head
            count = 0
            while itr:
                if count == index - 1:
                    itr.next = Node(data, itr.next)
                    break
                itr = itr.next
                count += 1
            
        print(f"Node {data} is inserted successfully at index {index}")

    def insert_after(self, value, data):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        is_inserted = False
        itr = self.head
        while itr:
            if itr.data == value:
                itr.next = Node(data, itr.next)
                is_inserted = True
                break
            itr = itr.next
            
        if is_inserted:
            print(f"Node {data} inserted after {value} successfully!")
        else:
            print(f"No match found for {value}, to insert node after!")
    
    def insert_before(self, value, data):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        is_inserted = False
        if self.head.data == value:
            self.insert_at_beginning(data, True)
            is_inserted = True
        elif self.get_length() > 1:
            itr = self.head
            while itr.next:
                if itr.next.data == value:
                    itr.next = Node(data, itr.next)
                    is_inserted = True
                    break
                itr = itr.next
                
        if is_inserted:
            print(f"Node {data} inserted before value {value} successfully!")
        else:
            print(f"No match found for {value}, to insert node before!")

    def first_node(self):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        print(f"Head node value: {self.head.data}")
        return self.head

    def last_node(self):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        itr = self.head
        while itr.next:
            itr = itr.next
        print(f"Last node value: {itr.data}")
        return itr

    def search(self, value):
        if self.head is None:
            print("Linked list is empty!")
            return False
        
        itr = self.head
        while itr:
            if itr.data == value:
                print(f"Value {value} found!")
                return True
            itr = itr.next
            
        print(f"Value {value} not found!")
        return False

    def delete_at_beginning(self, rotation: Optional[bool]=False):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        value = self.head.data
        self.head = self.head.next
        if not rotation:
            print(f"Node {value} at the beginning deleted!")
        return value

    def delete_at_end(self, rotation: Optional[bool]=False):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        value = None
        if self.get_length() == 1:
            value = self.head.data
            self.head = None
        
        if self.get_length() > 1:
            itr = self.head
            while itr.next.next:
                itr = itr.next
            value = itr.next.data
            itr.next = itr.next.next
            
        if not rotation:
            print(f"Node {value} at the end deleted!")
        return value
        
    def delete_at(self, index):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        if index < 0 or index >= self.get_length():
            raise Exception("Invalid Index")
        
        value = None
        if index == 0:
            value = self.delete_at_beginning(True)
        else:    
            count = 0
            itr = self.head
            while itr:
                if count == index - 1:
                    value = itr.next.data
                    itr.next = itr.next.next
                    break
                itr = itr.next
                count += 1
            
        print(f"Node {value} is deleted successfully at index {index}")
        return value

    def delete_by_value(self, value):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        is_deleted = False
        if self.head.data == value:
            self.head = self.head.next
            is_deleted = True
        
        if self.get_length() > 1:
            itr = self.head
            while itr.next:
                if itr.next.data == value:
                    itr.next = itr.next.next
                    is_deleted = True
                    break
                itr = itr.next
        
        if is_deleted:
            print(f"Node with value {value} deleted successfully!")
            return value
        else:
            print(f"No match found for {value}, to delete node by value!")

    def delete_after(self, value):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        data = None
        is_deleted = False
        itr = self.head
        while itr:
            if itr.data == value:
                if itr.next:
                    data = itr.next.data
                    itr.next = itr.next.next
                    is_deleted = True
                    break
                else:
                    print(f"No node exists after value {value} to delete!")
                    return
            itr = itr.next
                
        if is_deleted:
            print(f"Node {data} deleted after value {value} successfully!")
        else:
            print(f"No match found for {value}, to delete node after!")
            
        return data
    
    def delete_before(self, value):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        if self.head.data == value:
            print(f"No node to delete before the HEAD {self.head.data}!")
            return
        
        data = None
        is_deleted = False
        if self.get_length() > 1:            
            if self.head.next.data == value:
                data = self.delete_at_beginning(True)
                is_deleted = True
        
            if self.get_length() > 2:
                itr = self.head
                while itr.next.next:
                    if itr.next.next.data == value:
                        data = itr.next.data
                        itr.next = itr.next.next
                        is_deleted = True
                        break
                    itr = itr.next
            
        if is_deleted:
            print(f"Node {data} deleted before value {value} successfully!") 
        else:  
            print(f"No value {value} found to delete node before!")
            
        return data 
    
    def rotate_left(self, k):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        times = k%self.get_length()
        for _ in range(times):
            val  = self.delete_at_beginning(True)
            self.insert_at_end(val, True)
        
        print(f"The list rotated {k} times to the left successfully!")

    def rotate_right(self, k):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        times = k%self.get_length()
        for _ in range(times):
            val = self.delete_at_end(True)
            self.insert_at_beginning(val, True)
            
        print(f"The list rotated {k} times to the right successfully!")

    def sort(self):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        swapped = True
        while swapped:
            swapped = False
            itr = self.head
            while itr.next:
                if itr.data > itr.next.data:
                    itr.data, itr.next.data = itr.next.data, itr.data
                    swapped = True
                itr = itr.next
                
            if not swapped:
                break
            
        print("The list is sorted successfully!")

    def reverse(self):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        itr = self.head
        self.head = None
        while itr:
            self.insert_at_beginning(itr.data, True)
            itr = itr.next
            
        print("The list is reversed successfully!")
    
    def detect_loop(self):    
        if self.head is None:
            print("Linked list is empty!")
            return False
        
        container = set()
        itr = self.head
        while itr:
            if itr in container:
                print("Loop exists!")
                self.remove_loop(itr)
                return True
            container.add(itr)
            itr = itr.next
            
        print("No loop exist!")
        return False

    def floyd_cycle(self):
        if self.head is None:
            print("Linked list is empty!")
            return False
        
        slow_p = fast_p = self.head
        while slow_p and fast_p and fast_p.next:
            slow_p = slow_p.next
            fast_p = fast_p.next.next
            if slow_p == fast_p:
                self.remove_loop(slow_p)
                print("Loop exists!")
                return True
            
        print("No loop exist!")
        return False

    def remove_loop(self, itr):
        # approach - 1
        # ptr1 = ptr2 = itr
        # k = 1
        # while ptr1 != ptr2.next:
        #     ptr2 = ptr2.next
        #     k += 1
            
        # ptr1 = ptr2 = self.head
        # for _ in range(k):
        #     ptr2 = ptr2.next
            
        # while ptr1 != ptr2:
        #     ptr1 = ptr1.next
        #     ptr2 = ptr2.next
            
        # while ptr1 != ptr2.next:
        #     ptr2 = ptr2.next
        # ptr2.next = None
        # print("Loop removed successfully!")
        
        # approach - 2
        # p2 = itr
        # p1 = self.head
            
        # while p1 != p2:
        #   p2 = p2.next
        #   p1 = p1.next
            
        # while p1 != p2.next:
        #   p2 = p2.next
        # p2.next = None
        # print("Loop removed successfully!")
        
        # approach - 3
        p1 = p2 = itr
        while p1 != p2.next:
            p2 = p2.next
        p2.next = None
        print("Loop removed successfully!")
        
    def detect_loop_remove_loop_alternate(self):
        slow_p = fast_p = self.head
        while slow_p and fast_p and fast_p.next:
            slow_p = slow_p.next
            fast_p = fast_p.next.next
            if slow_p == fast_p:
                print("Loop exist!")
                slow_p = self.head
                while slow_p != fast_p:
                    slow_p = slow_p.next
                    fast_p = fast_p.next
            
                while slow_p != fast_p.next:
                    fast_p = fast_p.next
                fast_p.next = None
                print("Loop removed successfully!")
                return
        print("No loop exist!")
```
