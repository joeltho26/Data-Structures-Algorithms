```python
# Circular Doubly Linked list amortized - get_last_node, get_count, insert_at_end, insert_at_beginning operations
# rest of the operation comprises of same complexities

from typing import Optional

class Node:
    def __init__(self, data, prev=None, next=None):
        self.data = data
        self.next = next
        self.prev = prev
        
class LinkedList:
    def __init__(self):
        self.rear = self.head = None
        self.count = 0
        
    def insert_at_beginning(self, data, rotation: Optional[bool]=False):
        if self.head is None:
            self.rear = self.head = Node(data)
            self.rear.next = self.head.next = self.rear.prev = self.head.prev = self.head
            self.count += 1
        else:
            node = Node(data,next=self.head)
            self.head.prev = node
            self.head = node
            self.rear.next = node
            node.prev = self.rear
            self.count += 1
            
        if not rotation:
            print(f"Inserted node {data} at the beginning!")
    
    def insert_at_end(self, data, rotation: Optional[bool]=False):
        if self.head is None:
            self.rear = self.head = Node(data)
            self.rear.next = self.head.next = self.rear.prev = self.head.prev = self.head
            self.count += 1
        else:
            node = Node(data, prev=self.rear, next=self.rear.next)
            self.rear.next = node
            self.rear = self.head.prev = node
            self.count += 1
            
        if not rotation:
            print(f"Inserted node {data} at the end!")

    def print_forward(self):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        itr = self.head
        fstr = ""
        while itr != self.rear:
            fstr += str(itr.data) + "-->"
            itr = itr.next
        fstr += str(itr.data)
        print(f"Printing elements forward: \n {fstr}")
        print("#"*25)

    def get_length(self):
        return self.count
    
    def detect_loop_remove_loop_alternate(self):
        slow_p = fast_p = self.head
        while slow_p and fast_p and fast_p.next:
            slow_p = slow_p.next
            fast_p = fast_p.next.next
            if slow_p == fast_p:
                print("Loop exist!")
                slow_p = self.head
                while slow_p != fast_p:
                    slow_p = slow_p.next
                    fast_p = fast_p.next
            
                while slow_p != fast_p.next:
                    fast_p = fast_p.next
                fast_p.next = None
                print("Loops removed successfully!")
                return
        print("No loop exist!")
        
    def floyd_cycle(self):
        if self.head is None:
            print("Linked list is empty!")
            return False
        
        slow_p = fast_p = self.head
        while fast_p and fast_p.next:
            slow_p = slow_p.next
            fast_p = fast_p.next.next
            if slow_p == fast_p:
                #self.remove_loop(slow_p)
                print("Loop exists!")
                return True
            
        print("No loop exist!")
        return False
```

```python
if __name__ == "__main__":
    ll = LinkedList()
    ll.insert_at_end("Apple")
    ll.print_forward()
    ll.insert_at_beginning("Banana")
    ll.print_forward()
    ll.insert_at_end("Cherry")
    ll.print_forward()
    ll.insert_at_beginning("Lemon")
    ll.print_forward()
```

```python
ll.detect_loop_remove_loop_alternate()
ll.detect_loop_remove_loop_alternate()
ll.print_forward()
ll.floyd_cycle()
print("Count of items:", ll.get_length())
```
