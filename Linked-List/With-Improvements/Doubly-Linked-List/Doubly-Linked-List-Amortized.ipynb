```python
# Doubly Linked list amortized - get_last_node, get_count, insert_at_end operations
# rest of the operation comprises of same complexities

from typing import Optional 

class Node:
    def __init__(self, data, prev = None, next=None):
        self.data = data
        self.next = next
        self.prev = prev
        
class LinkedList:
    def __init__(self):
        self.rear = self.head = None
        self.count = 0
    
    def insert_at_end(self, data, rotation: Optional[bool]=False):
        if self.head is None:
            self.rear = self.head = Node(data)
            self.count += 1
        else:
            self.rear.next = Node(data, self.rear)
            self.rear = self.rear.next
            self.count += 1
            
        if not rotation:
            print(f"Node {data} inserted at the end successfully!")

    def print_forward(self):
        if self.head is None:
            print("Linked list is empty!")
            return
        
        itr = self.head
        fstr = ""
        while itr != self.rear:
            fstr += str(itr.data) + "-->"
            itr = itr.next
        fstr += str(itr.data)
        print(f"Printing elements forward: \n {fstr}")
        print("#"*25)
        
    def get_length(self):
        return self.count
    
    def get_last_node(self):
        print(f"Lastnode: {self.rear.data}")
        return self.rear
```

```python
if __name__ == "__main__":
    ll = LinkedList()
    ll.insert_at_end("Apple")
    ll.print_forward()
    ll.insert_at_end("Banana")
    ll.print_forward()
    ll.insert_at_end("Cherry")
    ll.print_forward()
    ll.insert_at_end("Lemon")
    ll.print_forward()
```

```python
print("Count of items:", ll.get_length())
print(ll.get_last_node())
```
