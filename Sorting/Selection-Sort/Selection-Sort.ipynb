```python
def selection_sort(arr):
    size = len(arr)
    for i in range(size):
        min_index = i
        for j in range(i+1,size):
            if arr[j] < arr[min_index]:
                min_index = j
        arr[i], arr[min_index] = arr[min_index], arr[i]
```

```python
if __name__ == "__main__":
    arr = [5, 3, 4, 1, 2, -4, 32.7, 12.6]
    selection_sort(arr)
    print(arr)
```
