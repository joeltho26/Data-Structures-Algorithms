```python
def bubble_sort(numbers):
    size = len(numbers)
    for i in range(size-1):
        swapped = False
        for j in range(i+1, size):
            if numbers[i] > numbers[j]:
                numbers[i], numbers[j] = numbers[j], numbers[i]
                swapped = True

        if not swapped:
            break
```

```python
numbers = [64, 34, 25, 12, 22, 11, 90]
bubble_sort(numbers)
print(numbers)
```
