```python
class Node:
    def __init__(self):
        self.children = [None] * 26
        self.is_end_of_word = False

class Trie():
    def insert(self, root, key):
        is_inserted = False
        current_node = root
        for c in key:
            index = ord(c)-ord("a")
            if not current_node.children[index]:
                is_inserted = True
                current_node.children[index] = Node()
            current_node = current_node.children[index]
        current_node.is_end_of_word = True
        
        if not is_inserted:
            print(f"'{key}' already available in the Trie!")
            return
        
        print(f"'{key}' is inserted into the Trie!")

    def search(self, root, key):
        current_node = root
        for c in key:
            index = ord(c)-ord("a")
            if not current_node.children[index]:
                print(f"'{key}' not found in Trie!")
                return False
            current_node = current_node.children[index]
            
        if current_node and current_node.is_end_of_word:
            print(f"'{key}' found in Trie!")
            return True
        else:
            print(f"'{key}' not found in Trie!")
            return False
    
    def is_empty(self, root):
        current_node = root
        for i in range(26):
            if current_node.children[i]:
                return False
        return True
    
    def delete(self, root, key, depth=0, is_found=None):
        if not root:
            return None
        
        if is_found is None:
            is_found = [False]
        
        if len(key) == depth:
            if root.is_end_of_word:
                is_found[0] = True
                root.is_end_of_word = False
                print(f"'{key}' found to delete!")

            if self.is_empty(root):
                del root
                root = None
            return root
        
        index = ord(key[depth]) - ord('a')
        root.children[index] = self.delete(root.children[index], key, depth+1, is_found)
        
        if self.is_empty(root) and not root.is_end_of_word:
            del root
            root = None
            
        if not is_found[0]:
            print(f"'{key}' not found to delete!")
            
        return root
```
